# Rating attributes on text.

<mark>**We recommend starting here.**</mark> GABRIEL can rate texts on arbitrary features (e.g. `"populism"`) to measure numerical scores on consistent 0-100 scales.

While classification is great for subsetting big datasets and big picture understanding, it loses a lot of the nuance. It is just yes or no on a label, not "how much".

Rating attributes answers that "how much" question. With `gabriel.rate` specify some attributes you want to measure, and **get numbers on a 0-100 scale for each attribute, corresponding to how much the text manifests that attribute**. This preserves nuance, and is great for a regression.

For example, say you have a dataset of speeches. Your attributes might be `"populism"` or `"critical of the supreme court"` or `"adversarial"` or anything. You'll get a rating for each speech, on each attribute, 0 to 100. Same if you had interviews with people about their childhood upbringing: you could rate `"emphasizes family over friends"` or any other features of interest.

Run the toy example below. **We recommend you try out `gabriel.rank` too** - it also generates ratings, but by comparing the text to each other rather that rating them abstractly, making it more grounded.

Ratings are as simple as `results = await gabriel.rate(your_data, column_name, attributes_to_rate, save_folder)`!

## Toy ratings example.

## Rating entities (e.g. 'Blackstone') instead of text.

GPT already knows a great deal about Rwanda, Nvidia, the spinning jenny, *Burger King v. Rudzewicz*, etc. It doesn't need text to rate - we can take advantage of its internal knowledge and web search capacity.

Say you want to know which state is more business friendly: Georgia or Wyoming. These are both *entities*, not texts. Or you want to know which Thanksgiving foods are the sweetest.

For many analyses, **acquiring text data is an unnecessary bottleneck; GPT's internal knowledge will suffice.**

You can pass entities (countries, companies, technologies, etc) via the `modality = "entity"` option. Everything else stays the same.

```python
toy_data = pd.DataFrame({"entity": [
    "turkey",
    "pumpkin pie",
    "pecan pie",
    "green bean casserole",
    "mashed potatoes",
    "cornbread",
    "paneer makhani",
]})

attributes = {
    "savory taste": "", # you can define an attribute in its value, or leave it blank
    "sweet taste": "",  # defining attributes helps anchor every measurement on different entities to use the same 'rubric'
    "bitter taste": "", # but GPT is quite consistent even without definitions
    "salty taste": "",  # you can use ChatGPT to help write out definitions
    "sour taste": "",
    "cornbready taste": "",
}

results = await gabriel.rate(
    toy_data,
    column_name = "entity",          # name of column
    attributes = attributes,         # attributes to score
    save_dir = "entity_rate",
    model = "gpt-5-mini",            # GPT model used for scoring
    n_runs = 3,
    modality = "entity",
    reset_files = False,
)
```

### Processing the speeches.

First we load our dataset, which consists of full text for all SotUs. This cell just pulls in a pre-cleaned dataset from the internet.

```python
from datasets import load_dataset
speeches = load_dataset("jsulz/state-of-the-union-addresses")
ds = speeches['train'].to_pandas()
```

Now, let's create small snippets from each speech to make it shorter. We'll get a sample of 30 speeches and create short paragraph-length snippets from each of those. This is simply for the purpose of making the example faster and cheaper to run - usually, you would not do this.

Run the below cell verbatim. This just loads small text snippets from each speech to get smaller examples to run. `target_words` sets the length of snippets - it is 500 words by default. **There is no need to change anything here, and you wouldn't generally do this for your data. This is just processing.**

```python
import random
import re
from bs4 import BeautifulSoup
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize, word_tokenize

# --- helpers ------------------------------------------------------

def strip_tags(html: str) -> str:
    """Return plain text with all tags removed."""
    return BeautifulSoup(html, "html.parser").get_text(" ", strip=True)

def sample_sentence_chunk(raw_html: str, target_words: int = 500) -> str:
    """
    Return a snippet that:
      • starts at a sentence boundary
      • ends at a sentence boundary
      • contains *at least* `target_words` words
    """
    text = strip_tags(raw_html)
    sentences = sent_tokenize(text)

    if not sentences:                       # defensive
        return ""

    word_counts = [len(word_tokenize(s)) for s in sentences]
    total_words = sum(word_counts)

    # If the whole speech is shorter than the budget, just return it.
    if total_words <= target_words:
        return text.strip()

    # Choose a random start sentence such that enough words remain.
    valid_starts = [
        i for i in range(len(sentences))
        if sum(word_counts[i:]) >= target_words
    ]
    start = random.choice(valid_starts)

    # Accumulate sentences until the budget is met or exceeded.
    words_so_far, end = 0, start
    while words_so_far < target_words and end < len(sentences):
        words_so_far += word_counts[end]
        end += 1

    snippet = " ".join(sentences[start:end])
    return snippet.strip()

# --- DataFrame wrapper --------------------------------------------

def get_snippets(df, html_col="speech_html", out_col="snippet",
                 target_words=500):
    """
    Add a column of sentence‑aligned snippets to `df`.
    """
    return df.assign(**{
        out_col: df[html_col].apply(
            lambda x: sample_sentence_chunk(x, target_words)
        )
    })

snippets = get_snippets(ds, html_col="speech_html", out_col="snippet",
                       target_words=500)

#The dataframe we are going to use, with 30 sample speeches each around 500 words in length
snippets_samp = snippets.sample(30, random_state = 42)
```

### Measuring our attributes.

# Ranking texts using pairwise comparisons.

Get grounded z scores as attribute ratings using a bunch of pairwise comparisons between texts.

Here we cover how to use `gabriel.rank` to get z scores for each text on how much it manifests each attribute relative to the other texts. We do this using pairwise comparisons many samples of two texts (or entities, images, audio; see multimodal section). Like a chess tournament, this creates ordered, ELO-like scores for each text. Unlike `gabriel.rate`, **these z scores are clearly grounded because they stem from comparisons**, not ratings in a vacuum.

The use case is similar to `gabriel.rate`: we want continuous numerical measurements on the texts. The difference here is our z scores will be **relative**, not absolute.

In this case, we do not pre-define a ratings scale like 0-100; we do many pairwise comparisons of texts over multiple rounds to get an ordering of where each text is in the relative distribution, separately for each attribute.

The `gabriel.rank` method call is analogous to `gabriel.rate`. `gabriel.rank` is more expensive and takes more time, since each entity is involved in many rounds of comparisons instead of one-shot ratings. Instead of `n_runs`, use the `n_rounds` and `matches_per_round` parameters to control how many iterations of the round-robin are run. More is better!

**We believe `gabriel.rate` is best for most use cases.** But if subtle relative differences and grounded ratings are needed, `gabriel.rank` is for you. Set `recursive` to `True` if you want identify the absolute top performers on an attribute in a large dataset!

## Toy rankings example.

Here's a simple exercise using rankings to measure a distribution of z scores across texts. We compare a series of short tweet-like texts to each other on a range of features.

```python
# Define the sample spreadsheet
toy_data = pd.DataFrame({"text": [
    "While your conclusion is careless, the deployment will likely improve consumer surplus.",
    "R&D intensity rose from 12% to 18% YoY; margins may compress before productivity gains materialize.",
    "That take is clown shoes—learn the basics before posting.",
    "Low-key hyped for on-device models—privacy win without the lag.",
    "@mina you’re right; the vendor ghosted us again—unacceptable.",
    "I concur with @lee: the policy update meaningfully strengthens user control.",
    "Subscription gating for security patches undermines trust.",
    "Empirical evidence suggests platform concentration exacerbates rent extraction absent interoperability mandates.",
    "New chips slap; battery life finally sane.",
    "Your deterministic framing is naïve; nonetheless, the architecture promises substantial welfare gains.",
    "If inference cost drops below $0.001/query, long-tail apps become viable; otherwise, unit economics fail.",
    "@devon is right—bait-and-switch trials are predatory; users deserve defaults that respect consent.",
    "Benchmarks improved 12→19 on MTEB; small but real.",
    "Anyone shouting “AI is over” is farming engagement—spare us.",
    "The deployment evidences credible privacy-by-design, a salutary precedent for the sector.",
    "Love the features, hate the upsells—pick a lane.",
    "Claims of AGI progress conflate scaling curves with capability thresholds; the evidence remains circumstantial.",
    "Good thread, @arya—clear, fair, helpful.",
    "Yes, lock-in is gross, but cross-vendor APIs mean switching costs are falling.",
    "With respect, your argument misrepresents the data and borders on bad faith."
]})

# Define the attributes we want to rate on
attributes = {
    "optimistic about technology": "Positive outlook on technology’s future or impact.",
    "anger": "Expression of frustration or hostility.",
    "formality": "Professional or academic style and structure.",
    "toxicity towards other users": "Insults, belittling, or demeaning remarks aimed at other users or groups.",
    "use of logic": "Explicit reasoning or evidence (numbers, if-then claims, comparisons, citations)."
}

# Run the call
results = await gabriel.rank(
    toy_data,
    column_name = "text",          # name of column with posts to rank
    attributes = attributes,       # attributes to rank by
    n_rounds = 3,                  # number of ELO rounds (more = more pairwise judgments)
    matches_per_round = 5,         # 5 * 3 means 15 total matchups for every entry
    save_dir = "toy_rank",         # directory to save results
    model = "gpt-5-mini",          # model used for pairwise comparisons
    modality = "text",             # input modality can also be "entity" (for terms like 'apple pie', not full texts), "pdf", "image", or "audio" (see multimodal section)
    n_attributes_per_run = 8,      # if more than 8 attributes, will split into separate calls of <= 8
    n_parallels = 650,             # max parallel threads (reduce if many errors, increase for higher speed)
    reset_files = False,  # rerunning the cell loads from save / picks up from checkpoint
    recursive = False,   # set to True to recursively cut low performers on a single attribute and rerank until revealing the cream of the crop (isolates and orders the true top performers)
)
```

# Creating taxonomies and discovering features, emergent from the data.

GABRIEL can compare thousands of texts from two class (e.g. 1-star and 5-star Glassdoor reviews) and discover features to explain the difference.

A great possibility of GPT is not just to measure features and test hypotheses you have in mind, but to discover novel explanatory features purely from the data.

We can leverage the sheer scale of GPT - have it **read thousands of documents** meticulously and many times over - to **find quantitative patterns** and uncover the most important trends. These patterns might be invisible to a human since they only reveal themselves at scale.

First we outline how `gabriel.bucket` can create a taxonomy out of thousands of entities (e.g. job tasks, technologies, court cases), learned purely from the data yet interpretable (unlike clustering).

Then we'll showcase `gabriel.discover`, which can ingest text/image/audio/entity data and tell you why one ad was persuasive while another was not.

## Taxonomies.

The aim of `gabriel.bucket` is to take a bunch of terms (potentially many thousands) and bucket them into mutually exclusive categories.

This is done by first giving GPT subsets of the term list, 250 terms per GPT call. If we have only a few hundred terms, we could just use ChatGPT to taxonomize them. But with thousands, one GPT can't attend to everything. Hence the splitting up.

This introduces a problem - we can get 10 mutually exclusive buckets for each set of 250 terms, but now we have hundreds of bucket candidates across these separate subsets!

We solve this by **voting**. We take all the bucket candidates from the prior round and redo the same activity - get GPT to look at 250 terms and come up with 10 buckets. But instead of generating the buckets, this time it can only vote for the 10 best buckets from the existing bucket candidates. We vote, and cull the bottom performers, and repeat until only the best candidates remain.

Lastly, to achieve mutual exclusivity in the final buckets: after many rounds of voting, we take the single bucket candidate with most votes and declare it one of our finalized buckets. Then we repeat voting, but only to fill the 9 other slots. Now, GPT must find buckets that compliment the one we have already finalized. Repeating this **results in a final, mutually exclusive bucket list**.

We can then classify the original terms on these top level buckets. We can **recursively repeat this whole process within each bucket class to get sub-buckets - and a full taxonomical tree** - if desired.

## Feature discovery.

`gabriel.discover` has two main modes. You could just pass one `column_name` (plus some `additional_instructions` on what you're looking for). For example, a column with a bunch of interview transcripts, and instructions that you are interested in the most common stressors in those people's lives. `gabriel.discover` will return a novel set of stressors - learned from the data - and what percent of the data exhibits each stressor.

Below we cover the other mode. Here you provide two different classes of the same data, one in `circle_column_name`, the other in `square_column_name`. These could be 1-star and 5-star Glassdoor reviews; Reddit arguments labelled persuasive and those which weren't (the example below); ad images from successful/unsuccessful campaigns; childhood biographies of upper class and middle class people. **The aim is to discover plain English features which explain the difference between your two classes.**

### Preparing the Reddit persuasion data.

```python
%pip install convokit
```

```python
from convokit import Corpus, download

winning = Corpus(filename=download("winning-args-corpus"))

winning_df = winning.get_utterances_dataframe()

convo_df = (
    winning.get_conversations_dataframe()
           .reset_index()                 # ✨ ‘id’ becomes a column
           .rename(columns={"id": "conversation_id",
                            "meta.op-userID": "op_user"})
           [["conversation_id", "op_user"]]   # keep just the two cols
)
```

```python
import pandas as pd
import re

# 1 ── split winners / losers & explode on pair_ids
winners = (
    winning_df[winning_df["meta.success"] == 1]
    .explode("meta.pair_ids")
    .merge(convo_df, on="conversation_id", how="left")
    # 1a  ensure the *winner* is not written by the OP
    .query("speaker != op_user")
)

losers = (
    winning_df[winning_df["meta.success"] == 0]
    .explode("meta.pair_ids")
)

# 2 ── join on the shared pair_id
paired = (
    winners.merge(losers,
                  on="meta.pair_ids",
                  suffixes=("_win", "_lose"))
    .rename(columns={"text_win":  "winning arguments",
                     "text_lose": "losing arguments"})
)

# 3 ── drop short or delta‑tainted texts
delta_regex   = re.compile(r"(?:Δ|∆|&amp;#8710;)", flags=re.IGNORECASE)

mask = (
    (paired["winning arguments"].str.len() >= 100) &
    (paired["losing arguments"].str.len()  >= 100) &
    ~(paired["winning arguments"].str.contains(delta_regex)) &
    ~(paired["losing arguments"].str.contains(delta_regex))
)

paired = paired.loc[mask]

# 4 ── keep the columns you want & sample 2 500 rows reproducibly
cols_keep = [
    "winning arguments", "losing arguments",
    "meta.pair_ids",                 # pairing key
    "conversation_id_win",           # CMV thread
    "speaker_win", "speaker_lose",
    "timestamp_win", "timestamp_lose"
]
paired = paired[cols_keep]

winning_to_run = paired.sample(n=2000, random_state=42).reset_index(drop=True)
```

### Running feature discovery to learn what makes an argument persuasive.

```python
results = await gabriel.discover(
    df = winning_to_run,
    circle_column_name = "winning arguments",
    square_column_name = "losing arguments",
    bucket_count = 12,
    additional_instructions = "What differentiates these arguments in terms of rhetorical style and strategy? Focus on how the arguments are written and the rhetoric used. Focus on specific differences in writing style, rhetorical devices, and rhetorical strategy.",
    save_dir = "reddit_persuasion",
    model = "gpt-5-mini",
    modality = "text",
    reset_files = False,
)
```

```python
sorted_results = results["summary"]
# in the label column of sorted_results, substitute substrings "circle entries" with "winning arguments" and "square entries" with "losing arguments"
sorted_results["label"] = sorted_results["label"].str.replace("circle entries", "winning arguments").str.replace("square entries", "losing arguments")

mask = sorted_results["net_pct"] < 0

# Swap "winning arguments" <-> "losing arguments" for negative rows
swapped = (sorted_results.loc[mask, "label"]
    .str.replace("winning arguments", "__tmp__", regex=False)
    .str.replace("losing arguments", "winning arguments", regex=False)
    .str.replace("__tmp__", "losing arguments", regex=False))

sorted_results.loc[mask, "label"] = swapped

# Flip sign of net_pct for those rows
sorted_results.loc[mask, "net_pct"] = -sorted_results.loc[mask, "net_pct"]

sorted_results = sorted_results.sort_values(by="net_pct", ascending=False)
```

# Deduplicating, filtering, and cleaning a large dataset.

GABRIEL has methods to deduplicate conceptually identical terms (company names, technologies, people), and to filter out irrelevant data.

## Deduplicating a large set of entities.

The `gabriel.deduplicate` method allows you to feed in thousands of entities and unify all conceptually identical terms into their single best representative term (e.g. `"turkey"`, `"thanksgiving turkey"`, and `"butterball"` map to `"turkey"`). You can tune how sensitive this process is just by asking, through the `additional_instructions` parameter.

Unlike simple fuzzy matching, this approach is intelligent. `gabriel.deduplicate` not only unifies all conceptually similar terms, but chooses the best term to map them to.

```python
technologies = [
    "iPhone", "Iphone 6", "iPhone 6s", "Apple smartphone", "F-16 Fighting Falcon", "F16",
    "GeForce RTX 4090", "GPU card", "bluetooth", "Bluetooth 5", "Li-ion battery", "Lithium ion cell",
    "Steam engine", "rotative steam-engine", "Electric Vehicle", "Tesla Model 3", "wireless earbuds", "AirPods",
    "photovoltaic panel", "solar pannel", "GPS", "satnav", "gps", "PS5", "PlayStation 5", "PlayStation",
    "Roomba", "MRI scanner", "quantum computer", "drone quadcopter", "Nintendo Switch", "hydraulic press",
]

tech_df = pd.DataFrame({"technology": technologies})

additional_instructions = """
Ensure all slight variants of the same technology are mapped to their best representative term from the list of technologies.
For example, "iPhone 4", "iPhone 6", "iPhone 15" etc all map to "iPhone".
All specific versions of a tech should be mapped to their generic tech name; avoid specificity.
"""

deduplicated_results = await gabriel.deduplicate(
    df = tech_df,
    column_name = "technology",
    save_dir = "toy_deduplication",
    model = "gpt-5-mini",
    n_runs = 3,             # repeats the process 3 times to be thorough
    additional_instructions = additional_instructions,
    use_embeddings = True,  # uses vector embeddings to deduplicate obvious matches and cluster similar terms
    group_size = 250,       # number of entities per GPT call, more is cheaper but lower performance
    reset_files = False,
)
```

## Filtering data based on a natural language condition.

`gabriel.filter` works like `gabriel.classify`: it returns True or False on a given condition. But `gabriel.filter` is designed for high throughout on massive datasets of terms (even millions).

While `gabriel.classify` assesses one text/image/audio/entity at a time on multiple labels, `gabriel.filter` assesses just one `condition` across hundreds of entities at a time. This is great for wrangling massive datasets and filtering to what you need.

For longer text data (not just terms like `"Marbury v. Madison"`) and smaller datasets, `gabriel.classify` is your best bet for filtering.

```python
athletes = [
    "Dan Marino", "Jennifer Lopez", "Allen Iverson", "Tom Cruise", "Andrew Luck", "Tom Brady", "Serena Williams", "Pete Davidson", "Lando Norris", "Rory McIlroy", "Usain Bolt",
    "Chris Paul", "Ryan Gosling", "Randy Moss", "Gordon Ramsay", "Katie Ledecky", "James Harden", "Patrick Ewing", "Zendaya", "Neymar", "Michael Phelps",
    "Conor McGregor", "Emma Stone", "Barry Sanders", "Sachin Tendulkar", "Abby Wambach", "Virat Kohli", "Tobey Maguire", "Simone Biles", "Fernando Alonso", "Billie Eilish",
    "Charles Barkley", "Tiger Woods", "Kylian Mbappé", "Leonardo DiCaprio", "Ariana Grande", "Dirk Nowitzki", "Mike Trout", "Hank Aaron"
]

athlete_df = pd.DataFrame({"athlete": athletes})

condition = """
The person must be an athlete who has won the top award in their sport (e.g Super Bowl, Olympic gold, etc) at least once.
Awards like MVP don't count; championships only.
Only include championship winning athletes.
"""

filtered_results = await gabriel.filter(
    df = athlete_df,
    column_name = "athlete",
    condition = condition,    # put your natural language condition here
    save_dir = "toy_filtering",
    model = "gpt-5-nano",
    entities_per_call = 100,  # number of terms per GPT call, higher is cheaper and works if only a small percentage of terms will meet the criteria
    shuffle = True,
    n_runs = 3,
    threshold = 0.25,         # repeats the process 3 times, and returns all terms which hit at least 25% of the time (at least once)
    reset_files = False,
)
```

# Deidentifying / anonymizing PII in text.

Using GPT for DeID tasks offers much better results than standard software, especially for social scientists.

Because it actually understands the content, GPT also can replace PII with fake sustitutes that maintain the original content's character and naturalness (replacing a name with an ethnically similar name, a hometown with a similarly sized different town, etc).

Crucially, **`gabriel.deidentify` can map all versions of a single PII entity** (`John Doe`, `Johnny`, `Jon Doe`, `J-O-H-N`) **to the same fake name** (`Brad Smith`). If there are multiple Johns, it will understand the different people and map them to separate fake names. These mappings will be consistent for a specific person across hundreds of pages of text (see second example).

Our approach is also designed to not unnecessarily mask famous people or places (e.g. Walmart), where de-identification is not required.

# Enforcing output diversity / creativity in GPT.

GABRIEL can force GPT to pursue distinct paths, or represent a population.

A common problem with using GPT at scale is that each instance of the model's outputs can be very similar to one another.

`gabriel.seed` addresses this by "seeding" a bunch of setups that are forced to be mutually exclusive. By creating all these seeds in a few prompts, GPT can ensure each seed is distinct from the others. The seeds can then be used to condition other prompts so outputs are distinct.

For example, `gabriel.seed` with `instructions` to suggest distinct writing styles, each seed with be a different writing style which can be used to condition a future prompt to different outcomes.

If `instructions` are to describe a set of people that match the US demographic distribution, the resulting seeds should be "people" who are not uniform (which normal repeated GPT prompting would get you) but instead accurately represent the US population across many axes.


```python
# you can instead just ask for mutual exclusivity and diversity of outputs, rather than representativeness
instructions = """
Each entity is two sentences describing core demographic traits of a person (age, sex, hometown, race, education level, occupation, etc).
The overall set must accurately represent the US population demographic distribution (including children) on all counts.
"""

people = await gabriel.seed(
    instructions = instructions,   # what sort of entity list we want
    num_entities = 1000,           # number of entities
    entities_per_generation = 50,  # entities per call (fewer is slower)
    entity_batch_frac = 0.25,      # fraction of overall dataset generated in one batch
    existing_entities_cap = 100,   # number of prior entities seen by each prompt
    save_dir = "toy_seed",
    model = "gpt-5.2",
    reset_files = False,
)
```

```python
people.sample(10)
```

```python
# now we can classify demographics to gut check that it is representative
labels = {
    "male person": "",
    "race is white": "",
    "race is asian": "",
    "californian": "",
    "college educated": "only if completed, not partial",
    "under 18": "",
}

demographics = await gabriel.classify(
    df = people,
    column_name = "entity",
    labels = labels,
    save_dir = "toy_seed_classify",
    model = "gpt-5-nano",
    reset_files = True,
)
```

```python
demographics.sample(5)
```

```python
label_cols = list(labels.keys())  # or whatever list of label column names you want

pct_true = (
    demographics[label_cols]
      .apply(lambda s: s.astype("boolean").fillna(False))  # handles True/False/None/"True"
      .mean() * 100
).sort_values(ascending=False)

print(pct_true.round(2))
```

> What if I have a custom prompt I want to run? I want to use GABRIEL's features but with my own tweaks!

We have at least three ways of accomodating this, from simple tweaks to full rewrites. Here we cover how you can use GABRIEL to run your own prompts and pipelines, while still taking advantage of our infrastructure.

## Option 1. Our prompts and pipelines, your guidelines (easiest).

The simplest way to edit the pipeline is to use our approaches off the shelf, but pass natural language clarifications through the **`additional_instructions`** parameter.

Whatever text is passed to the `additional_instructions` parameter simply gets appended to the end of our prompts. You can use it to clarify how ratings should be made; to provide labelled examples; to state that classification should only output one class, etc. The rest of our prompt and processing is unchanged. This parameter is available in all our functions: you can leverage the flexibility and intelligence of GPT by just telling it what you want or giving an example, if something isn't working as intended.

Below, see some examples of how we could use this parameter to clarify what GPT should do.
