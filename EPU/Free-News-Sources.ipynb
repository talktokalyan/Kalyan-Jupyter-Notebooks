```python
import internetarchive as ia  # pip install internetarchive

# Search for Indian newspapers
results = ia.search_items('collection:newspapers subject:"India" date:[2018 TO 2025]')
for r in results:
    print(r['identifier'])
```

```python
# pip install trafilatura
import trafilatura

url = "https://www.eenadu.net/telugu-news/world/pm-modi-speaks-with-president-dr-masoud-pezeshkian/0801/126050160"
downloaded = trafilatura.fetch_url(url)
text = trafilatura.extract(downloaded, include_comments=False, include_tables=False)
print(text)
```
