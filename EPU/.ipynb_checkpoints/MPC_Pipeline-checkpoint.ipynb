# RBI MPC Minutes Extraction Pipeline

**Single-cell notebook.** Run the cell below to:
1. Extract dates, rate, stance, member names, votes, and statements from all 56 PDFs
2. Export `mpc_minutes_extracted.xlsx` with three sheets:
   - `meeting_level` — one row per meeting (wide format)
   - `member_level_long` — one row per member per meeting
   - `extraction_log` — per-meeting quality flags


```python
r"""
RBI MPC Minutes Extraction Pipeline — v9
====================================================
"""

# ── Imports ────────────────────────────────────────────────────────────────────
import os, re, sys, time, logging
from datetime import date, datetime
from pathlib import Path
from typing import Optional

import pandas as pd
import pdfplumber
import openpyxl
import nltk
from nltk.tokenize import sent_tokenize
try:
    sent_tokenize("test sentence.")
except LookupError:
    nltk.download("punkt_tab", quiet=True)
from openpyxl.styles import Font, PatternFill, Alignment

# ── Logging ────────────────────────────────────────────────────────────────────
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s [%(levelname)s] %(message)s",
    handlers=[logging.StreamHandler(sys.stdout)],
    force=True,
)
for _lib in ("pdfminer", "pdfplumber", "PIL", "fontTools"):
    logging.getLogger(_lib).setLevel(logging.WARNING)
logger = logging.getLogger("mpc_pipeline")

# ── Configuration ──────────────────────────────────────────────────────────────
PDF_DIR     = os.path.join(os.path.expanduser("~"), "Documents", "Chetan Sir",
                           "NLP - Kalyan Sir", "Documents", "MPC-Minutes")
OUTPUT_PATH = os.path.join(os.path.expanduser("~"), "Documents", "Chetan Sir",
                           "NLP - Kalyan Sir", "mpc_minutes_extracted_v10.xlsx")

MIN_MEMBERS      = 5
EXPECTED_MEMBERS = 6
MAX_MEMBERS      = 6

MONTHS = {
    "january":1,"february":2,"march":3,"april":4,
    "may":5,"june":6,"july":7,"august":8,
    "september":9,"october":10,"november":11,"december":12,
}
NEW_SOURCE_LABEL_NO_RECORD = "inferred_no_mpc_record"

# ── PDF Reader ─────────────────────────────────────────────────────────────────

def extract_full_text(pdf_path: str) -> str:
    try:
        pages = []
        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                t = page.extract_text(x_tolerance=3, y_tolerance=3)
                pages.append(t or "")
        return _normalize_extracted_text("\n\n".join(pages))
    except Exception as e:
        logger.error(f"pdfplumber failed on {pdf_path}: {e}")
        return ""

# ── Utilities ──────────────────────────────────────────────────────────────────

def _clean(text: str) -> str:
    return re.sub(r'\s+', ' ', text).strip()

def _normalize_extracted_text(text: str) -> str:
    """Normalize PDF text while keeping useful section line breaks.
    
    Many RBI PDFs inject a standalone page number ('2', '3', etc.)
    on its own line between paragraphs during extraction.
    E.g. 'policy repo\\n\\n2\\nrate' should become 'policy repo rate'.
    """
    replacements = {
        "\u2018": "'", "\u2019": "'", "\u201c": '"', "\u201d": '"',
        "\u2013": "-", "\u2014": "-", "\u2212": "-", "\xa0": " ",
    }
    for src, dst in replacements.items():
        text = text.replace(src, dst)
    # Remove standalone single- or two-digit page numbers on their own line
    text = re.sub(r'\n\s*(\d{1,2})\s*\n', '\n', text)
    text = text.replace('\uf0b7', '•').replace("●", "•")
    text = re.sub(r'[ \t]+', ' ', text)
    text = re.sub(r'(?<=[A-Za-z])-\n(?=[a-z])', '', text)
    text = re.sub(r'\n{3,}', '\n\n', text)
    return text

def _normalize_for_regex(text: str) -> str:
    """Single-line normalized text for regexes that must survive PDF wraps."""
    return _clean(_normalize_extracted_text(text))

def _month_num(name: str) -> Optional[int]:
    return MONTHS.get(name.lower().strip())

def _fmt_date(d) -> str:
    if d is None:
        return ""
    return d.strftime("%d/%m/%Y") if isinstance(d, date) else str(d)

def _extract_sentence_around(text: str, pos: int, window: int = 500) -> str:
    """Find the sentence containing the regex match at `pos` using NLTK tokenizer.

    Falls back to a window heuristic only if NLTK fails to parse.
    """
    # Take a generous window around the match position
    ctx_start = max(0, pos - 200)
    ctx_end   = min(len(text), pos + window)
    context   = text[ctx_start:ctx_end]

    try:
        sentences = sent_tokenize(context)
    except Exception:
        return _clean(context)

    if not sentences:
        return _clean(context)

    # The match is at position (pos - ctx_start) within `context`.
    # Walk through sentence offsets to find the containing sentence.
    match_offset = pos - ctx_start
    running = 0
    best = sentences[0]
    for sent in sentences:
        sent_start = context.find(sent, running)
        if sent_start == -1:
            continue
        sent_end = sent_start + len(sent)
        if sent_start <= match_offset < sent_end:
            best = sent
            break
        running = sent_end

    return _clean(best)

def _sentences(text: str) -> list[str]:
    """Sentence tokenizer wrapper that always returns cleaned sentences."""
    text = _normalize_for_regex(text)
    if not text:
        return []
    try:
        return [_clean(s) for s in sent_tokenize(text) if _clean(s)]
    except Exception:
        return [s.strip() for s in re.split(r'(?<=[.!?])\s+', text) if s.strip()]

HONORIFIC_WORDS = {
    "dr", "shri", "smt", "prof", "professor", "ms", "mrs", "mr", "governor"
}

def _canonical_name(name: str) -> str:
    """Normalize member names for robust but auditable matching."""
    name = _normalize_for_regex(name).lower()
    name = re.sub(r'\([^)]*\)', ' ', name)
    name = re.sub(r'[^a-z0-9\s]', ' ', name)
    tokens = [t for t in name.split() if t not in HONORIFIC_WORDS]
    return " ".join(tokens)

def _name_tokens(name: str) -> list[str]:
    return _canonical_name(name).split()

def _last_name(name: str) -> str:
    toks = _name_tokens(name)
    return toks[-1] if toks else ""

def _name_match_score(target: str, candidate: str) -> int:
    """Higher score means stronger name match; 0 means no match."""
    t_can = _canonical_name(target)
    c_can = _canonical_name(candidate)
    if not t_can or not c_can:
        return 0
    if t_can == c_can:
        return 100
    t_tokens = t_can.split()
    c_tokens = c_can.split()
    if len(t_tokens) >= 2 and len(c_tokens) >= 2:
        if t_tokens[-1] == c_tokens[-1] and set(t_tokens).issubset(set(c_tokens) | {t_tokens[0]}):
            return 92
        if t_tokens[-1] == c_tokens[-1] and set(c_tokens).issubset(set(t_tokens) | {c_tokens[0]}):
            return 90
        shared = set(t_tokens) & set(c_tokens)
        if t_tokens[-1] == c_tokens[-1] and len(shared) >= 2:
            return 86
    if t_can in c_can or c_can in t_can:
        return 80
    surname = t_tokens[-1] if t_tokens else ""
    if len(surname) >= 3 and re.search(r'\b' + re.escape(surname) + r'\b', c_can):
        return 60
    return 0

def _best_match_key(name: str, mapping: dict):
    if not name or not mapping:
        return None
    scored = sorted(
        ((k, _name_match_score(name, str(k))) for k in mapping.keys()),
        key=lambda x: x[1],
        reverse=True,
    )
    if not scored or scored[0][1] <= 0:
        return None
    if len(scored) > 1 and scored[0][1] == scored[1][1] and scored[0][1] < 100:
        return None
    return scored[0][0]

def _member_names_in_text(text: str, members: list[dict]) -> list[str]:
    """Return attendance-list names mentioned in text, using exact then surname fallback."""
    text_can = _canonical_name(text)
    found = []
    for mem in members:
        name = mem.get("name", "")
        name_can = _canonical_name(name)
        if name_can and re.search(r'\b' + re.escape(name_can) + r'\b', text_can):
            found.append(name)
    if found:
        return found

    # Fallback: surname only, but only if it identifies a unique member.
    surname_to_names = {}
    for mem in members:
        surname_to_names.setdefault(_last_name(mem.get("name", "")), []).append(mem.get("name", ""))
    for surname, names in surname_to_names.items():
        if len(surname) >= 3 and len(names) == 1 and re.search(r'\b' + re.escape(surname) + r'\b', text_can):
            found.append(names[0])
    return list(dict.fromkeys(found))

def _extract_resolution_section(text: str) -> Optional[str]:
    """Return text of the Resolution section (bullet points + voting summary)."""
    m = re.search(r'(?:^|\n)\s*Resolution\s*\n', text, re.IGNORECASE)
    if not m:
        return None
    start = m.end()
    # End just before Assessment / Rationale / Growth and Inflation Outlook
    end_m = re.search(
        r'(?:^|\n)\s*(?:Assessment|Rationale for Monetary|Growth and Inflation Outlook)\s*\n',
        text[start:], re.IGNORECASE
    )
    return text[start: start + end_m.start()] if end_m else text[start: start + 2500]

def _extract_section(text: str, start_pattern: str, end_pattern: str) -> Optional[str]:
    sm = re.search(start_pattern, text, re.IGNORECASE)
    if not sm:
        return None
    em = re.search(end_pattern, text[sm.end():], re.IGNORECASE)
    return text[sm.end(): sm.end() + em.start()] if em else text[sm.end():]

# ── Date Parser ────────────────────────────────────────────────────────────────

def parse_dates(full_text: str, filename: str = "") -> dict:
    title_m = re.search(
        r'Minutes of the Monetary Policy Committee Meeting[,\s]*(.*?)(?:\n|\[Under)',
        full_text, re.IGNORECASE
    )
    candidate = _clean(title_m.group(1)) if title_m else ""
    held_m = re.search(
        r'was held (?:on |during |from )([A-Z][a-z]+.*?\d{4})',
        full_text, re.IGNORECASE
    )
    held_text = _clean(held_m.group(1)) if held_m else ""

    for raw, rule_prefix in [(candidate, "title"), (held_text, "held")]:
        if not raw:
            continue
        result = _parse_date_range(raw, rule_prefix)
        if result:
            result["candidate_text"] = raw
            return result

    logger.warning(f"[{filename}] Date extraction failed. title='{candidate}' held='{held_text}'")
    return {"start_date": None, "end_date": None, "matched_rule": "FAILED", "candidate_text": candidate}


def _parse_date_range(raw: str, prefix: str) -> Optional[dict]:
    raw = raw.strip().rstrip(',').strip()
    # A: "April 5-6, 2017"
    m = re.match(r'([A-Za-z]+)\s+(\d+)\s*[-–]\s*(\d+),?\s*(\d{4})', raw)
    if m:
        mn = _month_num(m.group(1))
        if mn:
            return {"start_date": date(int(m.group(4)), mn, int(m.group(2))),
                    "end_date":   date(int(m.group(4)), mn, int(m.group(3))),
                    "matched_rule": f"{prefix}_same_month_hyphen_range"}
    # B: "April 3 to 5, 2019"
    m = re.match(r'([A-Za-z]+)\s+(\d+)\s+to\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn = _month_num(m.group(1))
        if mn:
            return {"start_date": date(int(m.group(4)), mn, int(m.group(2))),
                    "end_date":   date(int(m.group(4)), mn, int(m.group(3))),
                    "matched_rule": f"{prefix}_same_month_range"}
    # C: "October 1, 3 and 4, 2019"
    m = re.match(r'([A-Za-z]+)\s+(\d+),\s*(\d+)\s+and\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn = _month_num(m.group(1))
        if mn:
            return {"start_date": date(int(m.group(5)), mn, int(m.group(2))),
                    "end_date":   date(int(m.group(5)), mn, int(m.group(4))),
                    "matched_rule": f"{prefix}_same_month_three_days"}
    # D: "May 2 and 4, 2022"
    m = re.match(r'([A-Za-z]+)\s+(\d+)\s+and\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn = _month_num(m.group(1))
        if mn:
            return {"start_date": date(int(m.group(4)), mn, int(m.group(2))),
                    "end_date":   date(int(m.group(4)), mn, int(m.group(3))),
                    "matched_rule": f"{prefix}_same_month_two_days"}
    # E1: "July 30-August 1, 2018"
    m = re.match(r'([A-Za-z]+)\s+(\d+)\s*[-–]\s*([A-Za-z]+)\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn1, mn2 = _month_num(m.group(1)), _month_num(m.group(3))
        if mn1 and mn2:
            return {"start_date": date(int(m.group(5)), mn1, int(m.group(2))),
                    "end_date":   date(int(m.group(5)), mn2, int(m.group(4))),
                    "matched_rule": f"{prefix}_cross_month_hyphen"}
    # E2: "September 29, 30 and October 1, 2025"
    m = re.match(r'([A-Za-z]+)\s+(\d+),\s*(\d+)\s+and\s+([A-Za-z]+)\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn1, mn2 = _month_num(m.group(1)), _month_num(m.group(4))
        if mn1 and mn2:
            return {"start_date": date(int(m.group(6)), mn1, int(m.group(2))),
                    "end_date":   date(int(m.group(6)), mn2, int(m.group(5))),
                    "matched_rule": f"{prefix}_cross_month_multiday"}
    # E3: "March 31, April 1 and 3, 2020"
    m = re.match(r'([A-Za-z]+)\s+(\d+),\s*([A-Za-z]+)\s+(\d+)(?:\s+and\s+(\d+))?,?\s*(\d{4})', raw)
    if m:
        mn1, mn2 = _month_num(m.group(1)), _month_num(m.group(3))
        if mn1 and mn2:
            last = int(m.group(5)) if m.group(5) else int(m.group(4))
            return {"start_date": date(int(m.group(6)), mn1, int(m.group(2))),
                    "end_date":   date(int(m.group(6)), mn2, last),
                    "matched_rule": f"{prefix}_cross_month"}
    # G: "June 4, 2020" – single day
    m = re.match(r'([A-Za-z]+)\s+(\d+),?\s*(\d{4})', raw)
    if m:
        mn = _month_num(m.group(1))
        if mn:
            d = date(int(m.group(3)), mn, int(m.group(2)))
            return {"start_date": d, "end_date": d, "matched_rule": f"{prefix}_single_day"}
    return None

# ── Member Parser ──────────────────────────────────────────────────────────────

_HONORIFIC_RE = re.compile(
    r'\band\s+(Dr\.?|Shri|Prof\.?|Professor|Ms\.?|Mrs\.?|Mr\.?)\s',
    re.IGNORECASE
)

def parse_members_from_attendance(full_text: str, filename: str = "") -> list[dict]:
    # Format (a): "attended by all the members – A; B; ... – and was chaired by [Gov]."
    m = re.search(
        r'(?:attended by all the members\s*[-–]\s*)(.*?)'
        r'\s*[-–]\s*and\s+was\s+chaired\s+by\s*(.*?(?:Governor|Chair).*?)\.',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        members = _parse_member_list(m.group(1), filename)
        gov = _parse_single_member(m.group(2), is_governor=True)
        if gov:
            members.append(gov)
        return members

    # Format (b): "chaired by [Gov] and was attended by all the members – A; B; ..."
    m = re.search(
        r'chaired by\s+(.*?(?:Governor|Chair).*?)\s+and\s+was\s+attended\s+by\s+all\s+the\s+members'
        r'\s*[-–]\s*(.*?)\.\s*(?=\n\s*\d+\.)',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        gov = _parse_single_member(m.group(1), is_governor=True)
        members = _parse_member_list(m.group(2), filename)
        if gov:
            members.append(gov)
        return members

    # Format (c): newer "chaired by…was attended by all the members – A; B" without look-ahead
    m = re.search(
        r'chaired by\s+(.*?(?:Governor|Chair).*?)\s+and\s+was\s+attended\s+by\s+all\s+the\s+members'
        r'\s*[-–]\s*(.*?)\.',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        gov = _parse_single_member(m.group(1), is_governor=True)
        members = _parse_member_list(m.group(2), filename)
        if gov:
            members.append(gov)
        if members:
            return members

    # Fallback
    chaired_m  = re.search(r'chaired by\s+([\w\.\s]+(?:Governor|Chair))', full_text, re.IGNORECASE)
    attended_m = re.search(
        r'members\s*[-–]\s*(.*?)(?:[-–]\s*and was chaired|\.\s*\d+\.)',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if attended_m:
        members = _parse_member_list(attended_m.group(1), filename)
        if chaired_m:
            gov = _parse_single_member(chaired_m.group(1), is_governor=True)
            if gov:
                members.append(gov)
        if members:
            return members

    logger.warning(f"[{filename}] Could not parse attendance paragraph")
    return []


def _parse_member_list(block: str, filename: str) -> list[dict]:
    block = _clean(block)
    parts = [p.strip() for p in re.split(r';\s*', block) if p.strip()]
    members = []
    for part in parts:
        part = re.sub(r'^(?:and|namely|also)\s+', '', part, flags=re.IGNORECASE).strip()
        inline = _HONORIFIC_RE.search(part)
        if inline:
            sub_parts = [
                part[:inline.start()].strip(),
                inline.group(1) + " " + part[inline.end():].strip(),
            ]
        else:
            sub_parts = [part]
        for sub in sub_parts:
            sub = re.sub(r'^(?:and|namely|also)\s+', '', sub, flags=re.IGNORECASE).strip()
            mem = _parse_single_member(sub, is_governor=False)
            if mem and len(mem["name"]) > 3:
                members.append(mem)
    return members


def _parse_single_member(text: str, is_governor: bool = False) -> Optional[dict]:
    text = _clean(text).strip(' –-')
    if not text:
        return None
    text = re.sub(r'\(the officer of[^)]+\)', '', text).strip()
    text = re.sub(r'\(the officer of[^)]+$',  '', text).strip()
    text = _clean(text)
    comma_idx = text.find(',')
    if comma_idx == -1:
        name        = text.strip()
        designation = "Governor" if is_governor else ""
    else:
        name        = text[:comma_idx].strip()
        designation = text[comma_idx + 1:].strip()
    if len(name) < 3:
        return None
    if not re.match(r'^(Dr\.?|Shri|Prof\.?|Professor|Ms\.?|Mrs\.?|Mr\.?|[A-Z])', name):
        return None
    return {"name": name, "designation": designation, "is_governor": is_governor}

# ── Rate Parser ────────────────────────────────────────────────────────────────

def parse_rate_decision(full_text: str, filename: str = "") -> dict:
    result = {"res_pol_rate_repo": None, "res_pol_rate_action": None, "matched_rate_sentence": None}
    resolution_text = _extract_resolution_section(full_text)
    search_text     = resolution_text if resolution_text else full_text[:4000]

    # 1. keep … unchanged at X per cent
    m = re.search(
        r'keep\s+the\s+policy\s+repo\s+rate\b.{0,200}?unchanged\s+at\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="unchanged",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # 1b. "voted to keep/maintain … policy repo rate … at X per cent" (newer format)
    # MUST come before the looser 1c to prevent matching the SDF/MSF rate instead.
    m = re.search(
        r'voted\b.{0,80}?(?:keep|maintain)\b.{0,150}?policy\s+repo\s+rate\b.{0,150}?'
        r'(?:unchanged\b.{0,30}?)?at\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="unchanged",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # 1c. policy repo rate … unchanged at X per cent (loose — can bleed to SDF rate)
    m = re.search(
        r'policy\s+repo\s+rate\b.{0,200}?\bunchanged\s+at\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="unchanged",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # 2. reduce … to Y per cent  (captures Y = new lower rate, not X = old rate)
    m = re.search(
        r'reduce\s+the\s+policy\s+repo\s+rate\b.{0,300}?to\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="cut",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # 2b. "voted to reduce … by X bps to Y per cent"  (newer format with no "reduce the")
    m = re.search(
        r'voted\s+to\s+reduce\s+the\s+policy\s+repo\s+rate\b.{0,300}?to\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="cut",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # 3. increase … to Y per cent
    m = re.search(
        r'increase\s+the\s+policy\s+repo\s+rate\b.{0,300}?to\s+([\d.]+)\s*per\s*cent',
        search_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="hike",
                      matched_rate_sentence=_extract_sentence_around(search_text, m.start()))
        return result

    # Fallback: voting table header
    m = re.search(
        r'Voting on the Resolution to\s+(keep|reduce|increase)\b.{0,100}?'
        r'(?:unchanged\s+)?at\s+([\d.]+)\s*per\s*cent',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        action_map = {"keep": "unchanged", "reduce": "cut", "increase": "hike"}
        result.update(res_pol_rate_repo=float(m.group(2)),
                      res_pol_rate_action=action_map.get(m.group(1).lower(), m.group(1).lower()),
                      matched_rate_sentence=_extract_sentence_around(full_text, m.start()))
        return result

    # Fallback 2: "reduce … to Y per cent" anywhere in text
    m = re.search(
        r'reduce\s+the\s+policy\s+repo\s+rate\b.{0,300}?to\s+([\d.]+)\s*per\s*cent',
        full_text, re.IGNORECASE | re.DOTALL
    )
    if m:
        result.update(res_pol_rate_repo=float(m.group(1)), res_pol_rate_action="cut",
                      matched_rate_sentence=_extract_sentence_around(full_text, m.start()))
        return result

    logger.warning(f"[{filename}] Rate decision extraction FAILED")
    return result

# ── Stance Parser ──────────────────────────────────────────────────────────────
#
# COMPLETE REWRITE to fix three categories of bugs:
#
#  Bug A  (Oct2018, Dec2018): "calibrated tightening" stance was absent from
#          the vocabulary. The parser fell through to full-text search and
#          matched "neutral" from a dissenting member's statement.
#          Fix: add calibrated_tightening as an explicit pattern searched
#          in the resolution section FIRST.
#
#  Bug B  (Oct2024, Jun2025): The resolution says "decided to change stance to
#          'neutral'" but the old pattern `neutral\s+stance` never matched
#          because the phrase "neutral stance" (adjective + noun adjacent) does
#          not appear — only "change stance to 'neutral'".
#          Fix: Priority-0 scan for "decided to change … stance … to [X]" that
#          extracts the new canonical stance before any keyword scan.
#
#  Bug C  (Apr2019, Dec2018): _determine_stance_action searched the combined
#          resolution + rationale text and matched "[Member] voted to change the
#          stance" from a dissenting vote summary, giving a false "change" action.
#          Fix: require the change language to be attributed to the MPC as a
#          body ("MPC decided to change", "also decided to change"), not an
#          individual member ("Dr. X voted to change").

# Canonical stance labels
_STANCE_CANONICAL = {
    "neutral":                    "neutral",
    "accommodative":              "accommodative",
    "withdrawal of accommodation":"withdrawal of accommodation",
    "calibrated tightening":      "calibrated tightening",
    "focused on withdrawal":      "withdrawal of accommodation",
}

def _normalise_stance(raw: str) -> Optional[str]:
    """Map a raw extracted phrase to a canonical stance label."""
    raw = raw.lower().strip(" '\"")
    for key, val in _STANCE_CANONICAL.items():
        if key in raw:
            return val
    return None


def parse_stance_decision(full_text: str, filename: str = "") -> dict:
    """
    Extract res_pol_stance, res_pol_stance_action, matched_stance_sentence.

    Priority order:
      0. Explicit "MPC decided to change stance from X to Y" → new stance is Y.
      1. "calibrated tightening" boilerplate in resolution section.
      2. "withdrawal of accommodation" keyword in resolution section.
      3. "accommodative stance" keyword in resolution section.
      4. "neutral stance" keyword in resolution section.
      5. Repeat 1-4 on full text (fallback).
    """
    result = {
        "res_pol_stance":        None,
        "res_pol_stance_action": None,
        "matched_stance_sentence": None,
        "stance_from_fallback":  False,
    }
    resolution_text = _extract_resolution_section(full_text)
    rationale_text  = _extract_section(full_text,
                                       r'\b(Outlook|Growth and Inflation Outlook|Rationale)\b',
                                       r'Voting on the Resolution')
    search_text = (resolution_text or "") + "\n" + (rationale_text or "")
    if not search_text.strip():
        search_text = full_text[:5000]

    # ── Priority 0: explicit "MPC / also decided to change stance … to [new]" ──
    # This handles Oct2024 ("change the monetary policy stance to 'neutral'")
    # and Jun2025 ("decided to change the stance from accommodative to neutral").
    CHANGE_TO_PATS = [
        # Match MPC-body decisions: 'decided to change stance to X'
        # Requires 'decided to' so that individual member dissent
        # ('Dr. X voted to change') is excluded.
        # Unicode curly quotes \u2018 \u2019 handle PDF typography.
        (
            r'decided\s+to\s+change\s+(?:the\s+)?(?:monetary\s+policy\s+)?'
            r'stance\s+(?:from\s+[\w\s\u2018\u2019\u201c\u201d\'"]+?\s+)?to\s+'
            r'[\u2018\u2019\u201c\u201d\'"]{0,1}(\w[\w ]*?)[\u2018\u2019\u201c\u201d\'"]{0,1}'
            r'(?:\s+and|\.|,|\n|$)'
        ),
    ]
    for pat in CHANGE_TO_PATS:
        m = re.search(pat, search_text, re.IGNORECASE | re.DOTALL)
        if m:
            raw_new = m.group(1).strip()
            canonical = _normalise_stance(raw_new)
            if canonical:
                result.update(
                    res_pol_stance=canonical,
                    res_pol_stance_action="change",
                    matched_stance_sentence=_extract_sentence_around(search_text, m.start()),
                )
                return result
            
    # ── Priority 0a: Hybrid "remain accommodative while focusing on withdrawal" ──
    # Used Apr2022 and May2022 only. Formal stance = accommodative;
    # withdrawal is a direction qualifier, not the stance label.
    hybrid_m = re.search(
        r'remain\s+accommodative\s+while\s+focusing\s+on\s+withdrawal\s+of\s+accommodation',
        search_text, re.IGNORECASE
    )
    if hybrid_m:
        result.update(
            res_pol_stance="accommodative",
            matched_stance_sentence=_extract_sentence_around(search_text, hybrid_m.start()),
            res_pol_stance_action=_determine_stance_action(
                resolution_text or search_text, "accommodative", filename
            ),
            stance_from_fallback=False,
        )
        return result
    
    # ── Priority 1-4: keyword scan ──────────────────────────────────────────────
    STANCE_PATTERNS = [
        # Note: calibrated tightening must be checked BEFORE neutral/accommodative
        # because some meeting boilerplates say "consistent with the stance of
        # calibrated tightening" which also contain the word "of" near "neutral".
        (r'\bcalibrated\s+tightening\b',             "calibrated tightening"),
        (r'(?:focused on |remain focused on )?withdrawal of accommodation',
                                                     "withdrawal of accommodation"),
        (r'\bstance\s+(?:at|as|to)\s+accommodative\b', "accommodative"),
        (r'\bstance\s+(?:at|as|to)\s+neutral\b',     "neutral"),
        (r'\baccommodative\s+stance\b',              "accommodative"),
        (r'\bneutral\s+stance\b',                    "neutral"),
        # Looser fallbacks for newer formats that say "remain accommodative"
        (r'\bremain\s+accommodative\b',              "accommodative"),
        (r'\bremain\s+neutral\b',                    "neutral"),
    ]

    for search_area, is_fallback in [(search_text, False), (full_text, True)]:
        for pattern, stance_label in STANCE_PATTERNS:
            m = re.search(pattern, search_area, re.IGNORECASE)
            if m:
                result.update(
                    res_pol_stance=stance_label,
                    matched_stance_sentence=_extract_sentence_around(search_area, m.start()),
                    res_pol_stance_action=_determine_stance_action(
                        search_text if not is_fallback else (resolution_text or search_text), stance_label, filename
                    ),
                    stance_from_fallback=is_fallback,
                )
                if is_fallback:
                    logger.warning(f"[{filename}] Stance '{stance_label}' found via full-text fallback, not from Resolution section")
                return result

    logger.warning(f"[{filename}] Stance extraction FAILED")
    return result


def _determine_stance_action(resolution_text: str, stance_label: str, filename: str) -> str:
    """
    v6 FIX (Bugs 1 & 2): CONTINUE patterns now evaluated BEFORE change patterns.
    Rationale: when a dissenter's sentence ("voted to change") appears in the same
    resolution block as the majority's "maintain/continue/consistent with" sentence,
    the majority signal must win. Checking continue first achieves this.
    """
    # ── STEP 1: CONTINUE signals (majority decision) ─────────────────────────
    CONTINUE_PATS = [
        r'maintain\s+the\s+stance\s+of',
        r'decided\s+to\s+(?:keep|maintain|retain|continue)\b.{0,80}?stance',
        r'continue\s+with\s+the',
        r'remain(?:ing)?\s+(?:focused\s+on|accommodative|neutral)',
        r'continue\s+(?:the\s+)?(?:accommodative|neutral)',
        r'retain(?:ing)?\s+(?:the\s+)?stance',
        r'maintain(?:ing)?\s+(?:the\s+)?(?:accommodative|neutral|calibrated)',
        r'consistent\s+with\s+the\s+stance\s+of',
        r'remain\s+accommodative\s+while\s+focusing\s+on\s+withdrawal',
        r'retain\s+the\s+stance\s+at',
    ]
    for pat in CONTINUE_PATS:
        if re.search(pat, resolution_text, re.IGNORECASE):
            return "continue"
 
    # ── STEP 2: Explicit unanimous-majority CHANGE for the adopted stance ─────
    # "voted in favour of changing" = named members supporting the change
    if re.search(
    r'voted\s+in\s+favou?r\s+of\s+(?:a\s+)?change\s+(?:in|of|to)\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance',
    resolution_text, re.IGNORECASE
    ):
        return "change"

    MPC_CHANGE_PATS = [
        r'(?:MPC|committee)\s+(?:also\s+)?decided\s+to\s+(?:change|shift)\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance',
        r'also\s+decided\s+to\s+(?:change|shift)\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance',
    ]
    MEMBER_VOTE_PAT = re.compile(
        r'(Dr\.|Shri|Prof\.|Professor|Ms\.|Mrs\.|Mr\.)\s+\w[\w\s]+\s+voted\s+to\s+(?:change|shift)\s+',
        re.IGNORECASE
    )
    for pat in MPC_CHANGE_PATS:
        for m in re.finditer(pat, resolution_text, re.IGNORECASE):
            context = resolution_text[max(0, m.start()-120): m.start()+20]
            if not MEMBER_VOTE_PAT.search(context):
                return "change"
 
    return "continue"   # safe default

# ── Voting Table Parser ────────────────────────────────────────────────────────

def parse_voting_table_details(full_text: str, filename: str = "") -> tuple[dict[str, str], dict[str, Optional[int]]]:
    """Extract rate votes and any explicit member-level bps magnitudes from the table."""
    header_m = re.search(r'Voting on the Resolution', full_text, re.IGNORECASE)
    if not header_m:
        logger.warning(f"[{filename}] Voting table not found")
        return {}, {}

    after = full_text[header_m.start():]
    votes = {}
    deltas = {}
    honorific = re.compile(r'^(Dr\.?|Shri|Prof\.?|Professor|Ms\.?|Mrs\.?|Mr\.?)\s*', re.IGNORECASE)
    header_context = _normalize_for_regex(after[:400])
    sign = -1 if re.search(r'\b(?:reduce|cut|lower)\b', header_context, re.IGNORECASE) else 1
    # Allow optional trailing number (magnitude of rate change column, e.g. "Yes 50")
    vote_line = re.compile(
        r'^(.*?)\s+(Yes|No)(?:\s+(-?\d{1,3})(?:\s*(?:bps|basis\s*points))?)?\s*$',
        re.IGNORECASE
    )

    for line in after.splitlines():
        line = line.strip()
        if not line:
            continue
        if re.match(r'^Statement by\b', line, re.IGNORECASE):
            break
        if not honorific.match(line):
            continue
        vm = vote_line.match(line)
        if vm:
            member = _clean(vm.group(1))
            votes[member] = vm.group(2).strip().capitalize()
            if vm.group(3):
                deltas[member] = sign * abs(int(vm.group(3)))

    if not votes:
        for line in after.splitlines()[:30]:
            line = line.strip()
            vm = re.match(r'^(.*?)\s+(Yes|No)(?:\s+(-?\d{1,3}))?\s*$', line, re.IGNORECASE)
            if vm and len(vm.group(1)) > 3:
                member = _clean(vm.group(1))
                votes[member] = vm.group(2).capitalize()
                if vm.group(3):
                    deltas[member] = sign * abs(int(vm.group(3)))

    if not votes:
        logger.warning(f"[{filename}] Voting table found but no votes extracted")
    return votes, deltas

def parse_voting_table(full_text: str, filename: str = "") -> dict[str, str]:
    """Backward-compatible wrapper: extract {member_name: Yes/No}."""
    return parse_voting_table_details(full_text, filename)[0]

# ── Stance Vote Parser ─────────────────────────────────────────────────────────

def parse_stance_votes(
    full_text: str, members: list[dict], rate_votes: dict, filename: str = "",
    statements: Optional[dict] = None, committee_stance: Optional[str] = None
) -> tuple[dict[str, str], dict]:
    """
    Return ({member_name: Yes/No}, metadata).

    Priority order:
      1. Explicit MPC-wide/all-member stance vote language.
      2. Member-specific sentence with vote/action verb and stance phrase.
      3. If partial explicit member coverage exists, infer remaining members
         as aligned with the MPC stance and flag those rows.
      4. Otherwise return no votes; never fabricate consensus from the rate table.
    """
    member_names = [m["name"] for m in members]
    votes: dict[str, str] = {}
    member_inferred: dict[str, bool] = {}
    sources: dict[str, str] = {}
    meta = {
        "stance_vote_unanimous_inferred": False,
        "stance_vote_member_inferred": False,
        "stance_vote_member_inferred_map": member_inferred,
        "stance_vote_sources": sources,
        "member_preferred_stance": {},
    }

    pre_statement_text = re.split(r'\bStatement\s+by\b', full_text, maxsplit=1, flags=re.IGNORECASE)[0]
    pre_statement_sentences = _sentences(pre_statement_text)

    # Priority 1: MPC-wide/all-member stance vote language.
    mpc_votes, mpc_meta = _extract_mpc_wide_stance_votes(pre_statement_text, members, committee_stance)
    votes.update(mpc_votes)
    sources.update(mpc_meta.get("sources", {}))
    meta["member_preferred_stance"].update(mpc_meta.get("preferred_stance", {}))
    if mpc_meta.get("unanimous"):
        meta["stance_vote_unanimous_inferred"] = True
        return votes, meta

    # Priority 2: member-specific statements; require vote/action verb and stance
    # phrase in the same sentence.
    statements = statements or {}
    for mem in members:
        name = mem.get("name", "")
        existing_source = sources.get(name, "")
        if name in votes and existing_source not in (
                    "mpc_wide_statement", "mpc_wide_reservation", "mpc_wide_exception",
                    "mpc_wide_all_except", "mpc_wide_member_named",
                    "mpc_partial_inference", "member_missing_inferred_from_mpc_stance",
                    "inferred_no_mpc_record"
                ):
                    continue
        stmt = _find_statement(name, statements)
        if not stmt:
            continue
        explicit, preferred = _extract_member_stance_vote_from_statement(stmt, committee_stance)
        if explicit:
            # For No-voters with no explicit preferred, try secondary inference before
            # flagging as missing_dissent_stance.
            if explicit == "No" and preferred is None:
                preferred = _infer_preferred_stance_from_statement(
                    _sentences(stmt), committee_stance
                )
            source = (
                "missing_dissent_stance"
                if explicit == "No" and preferred is None
                else "member_statement_explicit"
            )
            votes[name] = explicit
            sources[name] = source
            if preferred:
                meta["member_preferred_stance"][name] = preferred
        else:
            # Vote not overridden (mpc_wide stays). Still try to enrich preferred_stance
            # (e.g. Dec2023: vote=Yes via mpc_wide_reservation, but statement explicitly
            # says "it should be neutral").
            preferred_only = _infer_preferred_stance_from_statement(
                _sentences(stmt), committee_stance
            )
            if preferred_only and name not in meta["member_preferred_stance"]:
                meta["member_preferred_stance"][name] = preferred_only
            elif votes.get(name) == "No" and name not in meta["member_preferred_stance"]:
                # Existing No from mpc_wide (Aug2023/Apr2023/Oct2023 mpc_wide_reservation)
                # with no inferable preferred stance.
                sources[name] = "missing_dissent_stance"

    # Priority 3: partial explicit coverage -> infer remaining from the adopted
    # committee stance, while flagging each inferred member.
    if votes and len(votes) < len(member_names) and _has_stance_phrase(pre_statement_text):
        for name in member_names:
            if name not in votes:
                votes[name] = "Yes"
                member_inferred[name] = True
                sources[name] = "member_missing_inferred_from_mpc_stance"
                if committee_stance:
                    meta["member_preferred_stance"][name] = committee_stance
        meta["stance_vote_member_inferred"] = bool(member_inferred)
   # Priority 4: no MPC-wide votes at all — means no stance-vote narrative exists
    # (early meetings: Apr2017–Aug2018 approx). Individual statement votes (Priority 2)
    # are not evidence of a MPC-level narrative. Override any Priority-3 inferences.
    if not mpc_votes and committee_stance and rate_votes:
        has_stance_vote_language = bool(re.search(
            r'(?:voted|unanimously|expressed\s+reservations?|were\s+of\s+the\s+view)'
            r'.{0,260}(?:stance|withdrawal\s+of\s+accommodation)',
            pre_statement_text, re.IGNORECASE
        ))
        if not has_stance_vote_language:
            for name in member_names:
                if name not in votes or sources.get(name) == "member_missing_inferred_from_mpc_stance":
                    votes[name] = "Yes"
                    sources[name] = "inferred_no_mpc_record"
                    if committee_stance and name not in meta["member_preferred_stance"]:
                        meta["member_preferred_stance"][name] = committee_stance
                        # member_statement_explicit votes remain: Patel Oct2017 keeps his explicit src
            meta["stance_vote_unanimous_inferred"] = True

    return votes, meta



def _extract_mpc_wide_stance_votes(pre_statement_text: str, members: list[dict], committee_stance: Optional[str]) -> tuple[dict[str, str], dict]:
    text = _normalize_for_regex(pre_statement_text)
    sentences = _sentences(pre_statement_text)
    member_names = [m.get("name", "") for m in members]
    votes: dict[str, str] = {}
    sources: dict[str, str] = {}
    preferred: dict[str, str] = {}

    def set_vote(name: str, vote: str, source: str, pref: Optional[str] = None) -> None:
        votes[name] = vote
        sources[name] = source
        if pref:
            preferred[name] = pref
        elif vote == "Yes" and committee_stance:
            preferred[name] = committee_stance
        elif vote == "No":
            # Clear any previously-set preferred that simply echoes committee_stance.
            # A prior pass may have included this member in a majority group and set
            # preferred = committee_stance. A subsequent "exception/reservation" pass
            # correctly sets vote="No" — but must also invalidate that attribution.
            if name in preferred and preferred.get(name) == committee_stance:
                del preferred[name]
    # Explicit unanimous/all-member MPC stance decisions are high-confidence.
    if re.search(r'\bdecision\s+to\s+change\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance\s+was\s+unanimous\b', text, re.IGNORECASE):
        for name in member_names:
            set_vote(name, "Yes", "member_unanimous_mpc_statement")
        return votes, {"sources": sources, "preferred_stance": preferred, "unanimous": True}

    unanimous_change = re.search(
        r'\bunanimously\s+decided\b.{0,260}?\b(?:change|retain|continue|remain|keep)\b.{0,260}?(?:stance|withdrawal\s+of\s+accommodation)',
        text, re.IGNORECASE
    )
    if unanimous_change:
        for name in member_names:
            set_vote(name, "Yes", "member_unanimous_mpc_statement")
        return votes, {"sources": sources, "preferred_stance": preferred, "unanimous": True}

    for sent in sentences:
        if not _has_stance_phrase(sent):
            continue
        sent_norm = _normalize_for_regex(sent)
        has_action = re.search(r'\b(?:voted|decided|resolved|agreed)\b', sent_norm, re.IGNORECASE)
        unanimous = re.search(r'\bunanimously\b|\bdecision\s+to\s+change\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance\s+was\s+unanimous\b', sent_norm, re.IGNORECASE)
        all_member = re.search(r'\ball\s+members\b', sent_norm, re.IGNORECASE)
        if has_action and (unanimous or all_member):
            except_names = _extract_excepted_members(sent_norm, members)
            for name in member_names:
                set_vote(name, "Yes", "member_unanimous_mpc_statement" if not except_names else "mpc_wide_all_except")
            for name in except_names:
                set_vote(name, "No", "mpc_wide_exception", _opposed_stance_from_text(sent_norm, committee_stance))
            return votes, {"sources": sources, "preferred_stance": preferred, "unanimous": bool(not except_names)}
        if re.search(r'\bdecision\s+to\s+change\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance\s+was\s+unanimous\b', sent_norm, re.IGNORECASE):
            for name in member_names:
                set_vote(name, "Yes", "member_unanimous_mpc_statement")
            return votes, {"sources": sources, "preferred_stance": preferred, "unanimous": True}

    # Named majority support for the adopted stance. Find each stance-vote verb
    # and read the member list immediately before it; this survives "Dr."
    # abbreviations and PDF line breaks.
    for m in re.finditer(
        r'\bvoted\s+(?:in\s+favou?r\s+of\s+)?(?:to\s+)?'
        r'(?P<body>(?:for\s+)?(?:remain|remaining|continue|continuing|retain|retaining|keeping|keep|maintain|maintaining|change|changing|a\s+change|for\s+a\s+change)[^.]{0,260}?(?:stance|withdrawal\s+of\s+accommodation)[^.]{0,260})',
        text, re.IGNORECASE
    ):
        prefix = text[max(0, m.start() - 420):m.start()]
        prefix = re.split(r'\.\s+(?=(?:Dr|Prof|Professor|Shri|Ms|Mrs|Mr)\.?\s+)', prefix)[-1]
        prefix = re.split(r'\b(?:Regarding the stance|As regards the stance|As to the stance|On the stance)\b|;', prefix)[-1]
        names = _member_names_in_text(prefix, members)
        body = m.group("body")
        pref = _stance_label_in_text(body)
        vote = _stance_vote_relative_to_committee(body, committee_stance)
        # When voting YES on a "change FROM old TO committee_stance", pref is the
        # old stance — use committee_stance instead as the member's actual preference.
        if vote == "Yes" and pref and pref != committee_stance:
            pref = committee_stance
        source = "mpc_wide_member_named"
        for name in names:
            set_vote(name, vote, source, pref or committee_stance)

    # "However, two members - A and B, were of the view..." style dissent.
    for m in re.finditer(
        r'(?:however,\s*)?(?:two|three|four|five|six|\d+)\s+members?\s*-\s*(?P<names>.+?)\s*,?\s*'
        r'(?:were|was)\s+of\s+the\s+view\s+that\s+(?P<body>[^.]{0,260}?(?:stance|withdrawal\s+of\s+accommodation)[^.]{0,260})',
        text, re.IGNORECASE
    ):
        pref = _stance_label_in_text(m.group("body"))
        for name in _member_names_in_text(m.group("names"), members):
            set_vote(name, "No" if committee_stance and pref and pref != committee_stance else "Yes",
                     "mpc_wide_member_view", pref)

    if committee_stance and any(src == "mpc_wide_member_view" for src in sources.values()):
        for name in member_names:
            if name not in votes:
                set_vote(name, "Yes", "mpc_wide_majority_from_named_dissent", committee_stance)

    # Single-member "was of the view / retained his view" (Dec2025/Feb2026 Ram Singh style)
    for m in re.finditer(
        r'(?:Dr\.|Shri|Prof\.?|Professor|Ms\.|Mr\.)\s+(?P<names>[\w][\w\s\.]{2,40}?)\s+'
        r'(?:retain(?:s|ed)?\s+(?:his|her|their)\s+view|was\s+of\s+the\s+view|is\s+of\s+the\s+view)\s+'
        r'that\s+(?P<body>[^.]{0,260}?(?:stance|withdrawal\s+of\s+accommodation)[^.]{0,260})',
        text, re.IGNORECASE
    ):
        pref = _stance_label_in_text(m.group("body"))
        names = _member_names_in_text(m.group("names"), members)
        for name in names:
            vote = "No" if committee_stance and pref and pref != committee_stance else "Yes"
            set_vote(name, vote, "mpc_wide_statement", pref)

    # Reservations/did-not-support in a sentence often follows a named majority sentence.
    for sent in sentences:
        sent_norm = _normalize_for_regex(sent)
        if not re.search(r'\b(?:expressed\s+reservations?|did\s+not\s+(?:agree|support)|voted\s+against|dissented)\b', sent_norm, re.IGNORECASE):
            continue
        names = _member_names_in_text(sent_norm, members)
        if not names:
            continue
        for name in names:
            set_vote(name, "No", "mpc_wide_reservation", _opposed_stance_from_text(sent_norm, committee_stance))

    # If the MPC states a majority decision and only dissenters are named, keep
    # the dissent explicit and leave non-named members for low-confidence
    # inference below.
    return votes, {"sources": sources, "preferred_stance": preferred, "unanimous": False}

def _stance_vote_relative_to_committee(
    text: str,
    committee_stance: Optional[str]
) -> str:
    # NEW: If the body explicitly says "change/changing ... to [committee_stance]",
    # the member is SUPPORTING the change → Yes.
    # This handles Oct2024 "voted in favour of the change from withdrawal to neutral"
    # where body contains the OLD stance ("withdrawal") but the vote is for the NEW one.
    norm = _normalize_for_regex(text)
    # Positive support for transition TO committee stance
    if committee_stance and re.search(
        r'\b(?:vote|voted|support|supported|favo(?:u)?r|'
        r'favo(?:u)?red|agree|agreed|concur|concurred)\b'
        r'.{0,80}?\bto\s+[\'"]?' + re.escape(committee_stance) + r'[\'"]?',
        norm,
        re.IGNORECASE
    ):
        return "Yes"
    # F4: Caller may have stripped the leading "voted" verb when extracting body
    # (e.g. body = "for a change in stance from X to neutral"). Detect change/shift
    # targeting committee_stance even without a vote verb present.
    if committee_stance and re.search(
        r'\b(?:change|changing|shift|shifting)\b.{0,120}?\bto\s+[\'"]?'
        + re.escape(committee_stance) + r'[\'"]?',
        norm, re.IGNORECASE
    ):
        return "Yes"

    pref = _stance_label_in_text(text)

    # Explicit alternative stance detected
    if committee_stance and pref and pref != committee_stance:
        return "No"

    # Generic "change/shift" language without explicit target stance
    if (
        committee_stance
        and re.search(r'\b(?:change|changing|shift|shifting)\b', text, re.IGNORECASE)
        and pref != committee_stance
    ):
        return "No"

    return "Yes"

def _opposed_stance_from_text(text: str, committee_stance: Optional[str]) -> Optional[str]:
    pref = _stance_label_in_text(text)
    if pref and pref != committee_stance:
        return pref
    # v6 FIX (Bugs 9 & 10): When the committee stance is "withdrawal of accommodation"
    # and the member expressed reservations/voted against with NO explicit alternative,
    # their logically implied preference is "neutral" — the preceding stance.
    # This covers Varma/Goyal across all withdrawal-period meetings and
    # Aug2018 dissenters (Dholakia, Acharya, Patel).
    if committee_stance == "withdrawal of accommodation":
        if re.search(
            r'\bneutral\b|\bpause\b|not\s+withdraw|against\s+withdrawal',
            text,
            re.IGNORECASE
        ):
            return "neutral"
    # For calibrated tightening dissenters (Dec2018 Dholakia), they wanted neutral —
    # but _stance_label_in_text already extracts that from text, so we only reach here
    # in ambiguous cases. Return None for safety.
    return None

STANCE_PHRASE_RE = re.compile(
    r'\b(?:'
    r'(?:accommodative|neutral|calibrated\s+tightening)\s+(?:monetary\s+policy\s+)?stance|'
    r'(?:monetary\s+policy\s+)?stance\s+(?:of\s+|as\s+|being\s+)?(?:accommodative|neutral|calibrated\s+tightening)|'
    r'(?:monetary\s+policy\s+)?stance\s+(?:from\s+[\w\s\'\"]+?\s+)?to\s+(?:accommodative|neutral|calibrated\s+tightening)|'
    # v6 FIX (Bug 4): "support/change in the stance from X to Y" — Ram Singh Jun2025 style
    r'(?:support|change)\s+(?:in\s+)?(?:the\s+)?(?:monetary\s+policy\s+)?stance\s+from\s+[\w\s\'"]+?\s+to\s+[\w\s\'"]+|'
    r'withdrawal\s+of\s+accommodation|'
    r'withdrawal\s+of\s+the\s+accommodation|'
    r'withdraw(?:ing)?\s+accommodation'
    r')\b',
    re.IGNORECASE,
)

STANCE_ACTION_RE = re.compile(
    r'\b(?:I\s+)?(?:vote|voted|support|supported|concur|concurred|favo(?:u)?r|'
    r'favo(?:u)?red|agree|agreed|maintain|retain|continue|continuing|remain|'
    r'remaining|change|changing|shift|shifting|oppose|opposed)\b',
    re.IGNORECASE,
)

STANCE_NEGATIVE_RE = re.compile(
    r'\b(?:vote|voted)\s+against\b|'
    r'\bagainst\s+(?:this\s+)?(?:part\s+of\s+the\s+)?resolution\b|'
    r'\bexpress(?:ed|ing)?\s+reservations?\b|'
    r'\bhave\s+reservations?\b|'
    r'\bdissent(?:ed)?\b',
    re.IGNORECASE,
)

def _has_stance_phrase(text: str) -> bool:
    return bool(STANCE_PHRASE_RE.search(_normalize_for_regex(text)))

def _stance_label_in_text(text: str) -> Optional[str]:
    text = _normalize_for_regex(text)
    target_m = re.search(
    r'\bstance\b.{0,100}?\b(?:to|at|as)\s+(?:one\s+of\s+)?[\'"]?'
    r'(accommodative|neutral|calibrated\s+tightening|withdrawal\s+of\s+accommodation)[\'"]?',
    text, re.IGNORECASE
)
    if target_m:
        return _normalise_stance(target_m.group(1))
    for raw in [
        "withdrawal of accommodation",
        "withdrawal of the accommodation",
        "withdraw accommodation",
        "withdrawing accommodation",
        "accommodative stance",
        "neutral stance",
        "calibrated tightening",
        "accommodative",
        "neutral",
    ]:
        if re.search(r'\b' + re.escape(raw) + r'\b', text, re.IGNORECASE):
            return _normalise_stance(raw)
    return None

def _extract_excepted_members(sentence: str, members: list[dict]) -> list[str]:
    out = []
    for m in re.finditer(r'\bexcept\s+(.+?)(?:\s+-\s+|\s+voted\b|,?\s+who\b|$)', sentence, re.IGNORECASE):
        out.extend(_member_names_in_text(m.group(1), members))
    return list(dict.fromkeys(out))

def _extract_member_stance_vote_from_statement(statement: str, committee_stance: Optional[str] = None) -> tuple[Optional[str], Optional[str]]:
    sents = _sentences(statement)
    for i, sent in enumerate(sents):
        if not _has_stance_phrase(sent):
            # "I vote against this resolution also" often follows an explicit
            # "Turning to the stance..." sentence. Treat the second-resolution
            # construction as a formal stance dissent, not as commentary.
            prev_context = " ".join(sents[max(0, i - 2):i])
            if re.search(r'\bvote\s+against\s+(?:the\s+)?(?:second|this)\s+resolution\b', sent, re.IGNORECASE) and ( committee_stance or _has_stance_phrase(prev_context) or re.search(r'\bstance\b', prev_context, re.IGNORECASE)):
                return "No", None
            continue
        if not STANCE_ACTION_RE.search(sent):
            continue
        # For generic action verbs (remain/continue/maintain/retain/change/shift),
        # require first-person signal to distinguish member's vote from analysis.
        # "vote/support/favour/concur/oppose" are strong enough to stand alone.
        strong_vote_verbs = re.compile(
            r'\b(?:I\s+)?(?:vote|voted|favo(?:u)?r|favo(?:u)?red|concur|concurred|oppose|opposed)\b|'
            r'\bI\s+(?:support|agree|maintain|retain|continue|remain|change|shift)\b|'
            r'\bmy\s+vote\b',
            re.IGNORECASE,
        )
        if not strong_vote_verbs.search(sent) and not STANCE_NEGATIVE_RE.search(sent):
            continue  # skip analysis sentences with generic verbs and no first-person
        # Skip forward-looking sentences ("in the next meeting/policy/review")
        if re.search(
            r'\bnext\s+(?:MPC\s+)?(?:meeting|policy|review)\b',
            sent, re.IGNORECASE
        ):
            continue
        stance_label = _stance_label_in_text(sent)
        if STANCE_NEGATIVE_RE.search(sent):
            # "vote against X" means they OPPOSE X — X is not their preferred stance.
            if stance_label and committee_stance and stance_label == committee_stance:
                return "No", None
            return "No", stance_label
        if committee_stance and stance_label and stance_label != committee_stance:
            return "No", stance_label
        return "Yes", stance_label or committee_stance
    return None, None


def _name_in_text(name: str, text: str) -> bool:
    return bool(_member_names_in_text(text, [{"name": name}]))

# ── Statement Parser ───────────────────────────────────────────────────────────

def parse_statements(full_text: str, members: list[dict], filename: str = "") -> dict[str, str]:
    boundary_pattern = re.compile(r'(?:^|\n)\s*Statement\s+by\s+([^\n]+)', re.IGNORECASE)
    boundaries = list(boundary_pattern.finditer(full_text))
    if not boundaries:
        logger.warning(f"[{filename}] No 'Statement by' sections found")
        return {}

    statements = {}
    for i, bm in enumerate(boundaries):
        raw_name  = _clean(bm.group(1))
        # Occasionally the extracted heading keeps a paragraph number on the
        # same line. Keep only the heading-like prefix.
        raw_name = re.sub(r'\s+\d+\.\s+.*$', '', raw_name).strip()
        body_start = bm.end()
        body_end   = boundaries[i + 1].start() if i + 1 < len(boundaries) else len(full_text)
        body = full_text[body_start:body_end].strip()
        # Strip trailing Press Release / signature blocks
        body = re.sub(r'\n\(?\w+[\w\s]+\)?\nPress Release[^\n]*\n.*$', '', body, flags=re.DOTALL).strip()
        body = re.sub(r'\n[A-Z][a-z]+\s+[A-Z][a-z]+\s*\nPress Release.*$', '', body, flags=re.DOTALL).strip()
        body = _normalize_extracted_text(body).strip()
        if not body:
            logger.warning(f"[{filename}] Empty statement for '{raw_name}'")
        matched_names = _member_names_in_text(raw_name, members)
        key = matched_names[0] if len(matched_names) == 1 else raw_name
        if key in statements:
            logger.warning(f"[{filename}] Duplicate statement key after name matching: '{key}'")
            key = raw_name
        statements[key] = body

    if len(statements) != len(members):
        logger.warning(
            f"[{filename}] Found {len(statements)} statements but {len(members)} members. "
            f"Statement names: {list(statements.keys())}"
        )
    return statements

# ── Name-matching helpers ──────────────────────────────────────────────────────

def _find_vote(name: str, votes: dict) -> str:
    if not name or not votes:
        return ""
    key = _best_match_key(name, votes)
    if key is not None:
        return votes[key]
    return ""

def _find_statement(name: str, statements: dict) -> str:
    if not name or not statements:
        return ""
    key = _best_match_key(name, statements)
    if key is not None:
        return statements[key]
    return ""

# ── Member-Level Preferred Rate Parser ─────────────────────────────────────────

RATE_ACTION_RE = re.compile(
    r'\b(?:vote|voted|support|supported|favo(?:u)?r|favo(?:u)?red|prefer|preferred)\b',
    re.IGNORECASE,
)
RATE_TARGET_RE = re.compile(r'\b(?:policy\s+repo\s+rate|repo\s+rate|policy\s+rate)\b', re.IGNORECASE)
FORMAL_RATE_VOTE_RE = re.compile(
    r'\b(?:I\s+)?(?:vote|voted|prefer|preferred|favo(?:u)?r|favo(?:u)?red|plead|urge|advocate)\b|'
    r'\bI\s+(?:support|supported|concur|concurred|agree|agreed)\b|'
    r'\bmy\s+vote\s+(?:is|was)\b',
    re.IGNORECASE,
)

def parse_member_preferred_deltas(statements: dict, members: list[dict], filename: str = "") -> dict[str, int]:
    """Extract member-specific preferred repo-rate deltas from statements."""
    deltas = {}
    for mem in members:
        name = mem.get("name", "")
        stmt = _find_statement(name, statements)
        if not stmt:
            continue
        delta = _extract_preferred_delta_from_statement(stmt)
        if delta is not None:
            deltas[name] = delta

    return deltas

def parse_mpc_narrative_preferred_deltas(full_text: str, members: list[dict], filename: str = "") -> dict[str, int]:
    """Extract member-specific rate preferences from MPC-wide narrative text."""
    pre_statement_text = re.split(r'\bStatement\s+by\b', full_text, maxsplit=1, flags=re.IGNORECASE)[0]
    text = _normalize_for_regex(pre_statement_text)
    deltas = {}
    for m in re.finditer(r'\bvoted\b(?P<body>[^.;]{0,220}?(?:basis\s+points?|bps|bp|status\s+quo|unchanged)[^.;]{0,100})', text, re.IGNORECASE):
        prefix = text[max(0, m.start() - 320):m.start()]
        prefix = re.split(r'\.\s+(?=(?:Dr|Prof|Professor|Shri|Ms|Mrs|Mr)\.?\s+)', prefix)[-1]
        prefix = re.split(r'\bvoted\b', prefix, flags=re.IGNORECASE)[-1]
        prefix = re.split(r'\bwhile\b|;', prefix)[-1]
        names = _member_names_in_text(prefix, members)
        delta = _extract_rate_delta_from_sentence("voted " + m.group("body"))
        if names and delta is not None:
            for name in names:
                deltas[name] = delta
    for sent in _sentences(pre_statement_text):
        if not re.search(r'\bvoted\b', sent, re.IGNORECASE):
            continue
        if not re.search(r'\b(?:policy\s+repo\s+rate|repo\s+rate|policy\s+rate|rate|status\s+quo|unchanged|basis\s+points?|bps)\b', sent, re.IGNORECASE):
            continue
        for m in re.finditer(r'\bvoted\b(?P<body>[^.;]{0,180}?(?:basis\s+points?|bps|bp|status\s+quo|unchanged)[^.;]{0,80})', sent, re.IGNORECASE):
            prefix = sent[max(0, m.start() - 260):m.start()]
            prefix = re.split(r'\.\s+(?=(?:Dr|Prof|Professor|Shri|Ms|Mrs|Mr)\.?\s+)', prefix)[-1]
            prefix = re.split(r'\bvoted\b', prefix, flags=re.IGNORECASE)[-1]
            prefix = re.split(r'\bwhile\b|;', prefix)[-1]
            names = _member_names_in_text(prefix, members)
            if not names:
                continue
            delta = _extract_rate_delta_from_sentence("voted " + m.group("body"))
            if delta is None:
                continue
            for name in names:
                deltas[name] = delta
    # Catch "unanimously decided to reduce/increase...by X bps" — no "voted" keyword
    # so the loop above misses it entirely (Jun2019: "unanimously decided to reduce
    # the policy repo rate by 25 basis points"). Since mpc_deltas overrides stmt_deltas
    # in combined_deltas, this correctly fixes Acharya's -50 (from statement) to -25.
    if not deltas:
        unanim_m = re.search(
            r'\bunanimously\s+decided\s+to\s+(?P<action>reduce|cut|increase|raise|hike)\b'
            r'[^.;]{0,200}?(\d{1,3})\s*(?:basis\s*points?|bps|bp)\b',
            text, re.IGNORECASE
        )
        if unanim_m:
            action = unanim_m.group("action").lower()
            sign = -1 if action in ("reduce", "cut") else 1
            unm_delta = sign * int(unanim_m.group(2))
            for mem in members:
                deltas[mem.get("name", "")] = unm_delta

    return deltas

def _extract_preferred_delta_from_statement(statement: str) -> Optional[int]:
    # F7: past-tense / prior-meeting reference patterns. Members occasionally
    # quote their previous vote ("In the October 2025 MPC minutes, I had noted
    # ... which had prompted me to vote for a pause"). Without this guard, the
    # historical sentence is matched and returns 0 before the actual current
    # vote sentence is reached.
    HISTORICAL_RE = re.compile(
        r'\bhad\s+(?:noted|argued|voted|preferred|favo(?:u)?red|supported|pleaded|recommended)\b|'
        r'\bin\s+(?:the\s+)?(?:January|February|March|April|May|June|July|August|'
        r'September|October|November|December)\s+\d{4}\s+(?:MPC|meeting|minutes|policy)\b|'
        r'\b(?:previously|earlier|last\s+(?:meeting|policy|MPC))\b',
        re.IGNORECASE,
    )
    for sent in _sentences(statement):
        if HISTORICAL_RE.search(sent):
            continue
        if not FORMAL_RATE_VOTE_RE.search(sent):
            continue
        sent_norm = _normalize_for_regex(sent)
        delta = _extract_rate_delta_from_sentence(sent_norm)
        if delta is not None:
            return delta
    return None

def _extract_rate_delta_from_sentence(sentence: str) -> Optional[int]:
    sent_norm = _normalize_for_regex(sentence)
    has_status_quo = re.search(r'\b(?:pause|status\s+quo|unchanged|no\s+change)\b', sent_norm, re.IGNORECASE)
    if not (RATE_TARGET_RE.search(sent_norm) or re.search(r'\b(?:basis\s*points?|bps|bp)\b', sent_norm, re.IGNORECASE) or has_status_quo):
        return None
        # Fix 8a: add plead/urge/advocate to formal_pos

    formal_pos = re.search(
        r'\b(?:I\s+)?(?:vote|voted|prefer|preferred|favo(?:u)?r|favo(?:u)?red|plead|urge|advocate)\b|'
        r'\bI\s+(?:support|supported|concur|concurred|agree|agreed)\b|'
        r'\bmy\s+vote\s+(?:is|was)\b',
        sent_norm, re.IGNORECASE
    )
    if not formal_pos:
        return None
    formal_tail = sent_norm[formal_pos.start():formal_pos.start() + 220]

    m = re.search(
        r'\b(?:increase|raise|raised|raising|hike|rise)\b.{0,120}?'
        r'(\d{1,3})[\s\-]*(?:basis\s*points?|bps|bp)\b',
        formal_tail, re.IGNORECASE
    )
    if m:
        return int(m.group(1))

    m = re.search(
        r'\b(?:reduce|cut|lower|lowered|reduction)\b.{0,120}?'
        r'(\d{1,3})[\s\-]*(?:basis\s*points?|bps|bp)\b',
        formal_tail, re.IGNORECASE
    )
    if m:
        return -int(m.group(1))    # ← this was missing in your paste

    m = re.search(
        r'(\d{1,3})[\s\-]*(?:basis\s*points?|bps|bp)\b.{0,80}?'
        r'\b(?:increase|raise|rise|hike)\b',
        formal_tail, re.IGNORECASE
    )
    if m:
        return int(m.group(1))

    m = re.search(
        r'(\d{1,3})[\s\-]*(?:basis\s*points?|bps|bp)\b.{0,80}?'
        r'\b(?:reduce|cut|lower|reduction)\b',    # catches "50 basis points cut" or "25-bps cut"
        formal_tail, re.IGNORECASE
    )
    if m:
        return -int(m.group(1))

    if re.search(r'\b(?:pause|status\s+quo|unchanged|no\s+change)\b', formal_tail, re.IGNORECASE) and (
        RATE_TARGET_RE.search(sent_norm) or
        re.search(r'\b(?:vote|voted)\s+for\s+(?:a\s+)?(?:pause|status\s+quo)\b',
                formal_tail, re.IGNORECASE)
    ):
        return 0

    return None

def _committee_delta_bps(rec: dict, prev_rate: Optional[float]) -> Optional[int]:
    repo = rec.get("res_pol_rate_repo")
    if repo is None or prev_rate is None:
        return None
    return int(round((float(repo) - float(prev_rate)) * 100))

_PRE_MPC_REPO_RATE: float = 6.25

def _previous_repo_by_file(records: list[dict]) -> dict[str, Optional[float]]:
    """v6 FIX (Bug 12): seed prev_rate with the pre-MPC rate so the first meeting
    gets a valid previous_repo_rate instead of None/0."""
    prev_by_file = {}
    prev_rate: Optional[float] = _PRE_MPC_REPO_RATE   # ← changed from None
    for rec in sorted(records, key=lambda r: r.get("start_date") or date(1900, 1, 1)):
        fname = rec.get("file_name", "")
        prev_by_file[fname] = prev_rate
        if rec.get("res_pol_rate_repo") is not None:
            prev_rate = rec.get("res_pol_rate_repo")
    return prev_by_file

def _member_preferred_rate_fields(name: str, rec: dict, prev_rate: Optional[float]) -> dict:
    rate_vote = _find_vote(name, rec.get("rate_votes", {}))
    table_delta = _find_vote(name, rec.get("rate_vote_deltas", {}))
    stmt_delta = _find_vote(name, rec.get("member_preferred_deltas", {}))
    delta_source_map = rec.get("member_preferred_delta_sources", {})
    committee_delta = _committee_delta_bps(rec, prev_rate)

    delta = None
    source = ""
    review = False
    if table_delta not in ("", None):
        delta = int(table_delta)
        source = "voting_table_explicit"
    elif stmt_delta not in ("", None):
        delta = int(stmt_delta)
        source = _find_vote(name, delta_source_map) or "member_statement_explicit"
    elif rate_vote == "Yes" and committee_delta is not None:
        delta = committee_delta
        source = "committee_resolution_yes_vote"
    if rate_vote == "No" and delta is None:
        review = True
        source = "missing_dissent_magnitude"
    elif delta is None and prev_rate is None:
        review = True
        source = "missing_previous_repo_rate"

    implied = None
    if delta is not None and prev_rate is not None:
        implied = round(float(prev_rate) + delta / 100.0, 2)
    elif delta is not None:
        review = True
        source = source or "missing_previous_repo_rate"

    return {
        "previous_repo_rate": prev_rate,
        "member_preferred_delta_bps": delta,
        "member_implied_repo_rate": implied,
        "member_rate_vote_source": source,
        "flag_member_rate_review": int(review),
    }

# ── Full Pipeline for One PDF ──────────────────────────────────────────────────

def process_pdf(pdf_path: str) -> dict:
    filename = os.path.basename(pdf_path)
    logger.info(f"Processing: {filename}")

    record = {
        "file_name": filename, "source_path": pdf_path,
        "start_date": None, "end_date": None, "matched_rule": None, "candidate_text": None,
        "res_pol_rate_repo": None, "res_pol_rate_action": None,
        "res_pol_stance": None, "res_pol_stance_action": None,
        "matched_rate_sentence": None, "matched_stance_sentence": None,
        "members": [], "rate_votes": {}, "stance_votes": {},
        "rate_vote_deltas": {}, "member_preferred_deltas": {},
        "stance_vote_inferred": False, "stance_vote_unanimous_inferred": False,
        "stance_vote_member_inferred": False, "stance_vote_member_inferred_map": {},
        "stance_vote_sources": {}, "member_preferred_stance": {},
        "stance_from_fallback": False,
        "statements": {}, "issues_summary": [],
    }

    try:
        full_text = extract_full_text(pdf_path)
    except Exception as e:
        logger.error(f"[{filename}] Text extraction crashed: {e}")
        return record

    if not full_text.strip():
        logger.error(f"[{filename}] Empty text extracted")
        return record

    try:
        record.update(parse_dates(full_text, filename))
    except Exception as e:
        logger.error(f"[{filename}] Date parsing crashed: {e}")

    try:
        record["members"] = parse_members_from_attendance(full_text, filename)
    except Exception as e:
        logger.error(f"[{filename}] Member parsing crashed: {e}")

    try:
        record.update(parse_rate_decision(full_text, filename))
    except Exception as e:
        logger.error(f"[{filename}] Rate parsing crashed: {e}")

    try:
        record.update(parse_stance_decision(full_text, filename))
    except Exception as e:
        logger.error(f"[{filename}] Stance parsing crashed: {e}")

    try:
        rv, rv_deltas = parse_voting_table_details(full_text, filename)
        record["rate_votes"] = rv
        record["rate_vote_deltas"] = rv_deltas
    except Exception as e:
        logger.error(f"[{filename}] Voting table parsing crashed: {e}")
    # Post-process: infer rate_vote=No when member's table delta ≠ committee delta.
    # Rationale: magnitude divergence is effective dissent (Oct2022 Ashima precedent).
    try:
        rv_deltas = record.get("rate_vote_deltas", {})
        if rv_deltas and record.get("members"):
            for name, tbl_delta in rv_deltas.items():
                if tbl_delta is None:
                    continue
                # Find committee delta from the rate decision
                # We don't have prev_rate yet at this stage; use action as proxy
                if record["rate_votes"].get(name) == "Yes":
                    # Check if any other member has a DIFFERENT explicit delta
                    all_deltas = set(v for v in rv_deltas.values() if v is not None)
                    if len(all_deltas) > 1:
                        # Heterogeneous table → find modal (most common) delta
                        from collections import Counter
                        modal_delta = Counter(v for v in rv_deltas.values() if v is not None).most_common(1)[0][0]
                        if tbl_delta != modal_delta:
                            record["rate_votes"][name] = "No"
                            logger.info(f"[{filename}] {name}: rate_vote→No (table delta {tbl_delta} ≠ modal {modal_delta})")
    except Exception as e:
        logger.error(f"[{filename}] Rate vote magnitude adjustment crashed: {e}")

    try:
        record["statements"] = parse_statements(full_text, record["members"], filename)
    except Exception as e:
        logger.error(f"[{filename}] Statement parsing crashed: {e}")

    try:
        mpc_deltas = parse_mpc_narrative_preferred_deltas(full_text, record["members"], filename)
        stmt_deltas = parse_member_preferred_deltas(
            record["statements"], record["members"], filename
        )
        combined_deltas = dict(stmt_deltas)
        combined_deltas.update(mpc_deltas)  # MPC narrative has higher precedence when explicit.
        record["member_preferred_deltas"] = combined_deltas
        record["member_preferred_delta_sources"] = {
            **{k: "member_statement_explicit" for k in stmt_deltas},
            **{k: "mpc_wide_rate_statement" for k in mpc_deltas},
        }
    except Exception as e:
        logger.error(f"[{filename}] Member preferred-rate parsing crashed: {e}")

    try:
        sv, meta = parse_stance_votes(
            full_text, record["members"], record["rate_votes"], filename,
            record["statements"], record.get("res_pol_stance")
        )
        record["stance_votes"] = sv
        record.update(meta)
        record["stance_vote_inferred"] = bool(
            record.get("stance_vote_unanimous_inferred") or record.get("stance_vote_member_inferred")
        )
    except Exception as e:
        logger.error(f"[{filename}] Stance vote parsing crashed: {e}")

    try:
        record["semantic_validation_flags"] = _semantic_validation_flags(full_text, record)
    except Exception as e:
        logger.error(f"[{filename}] Semantic validation crashed: {e}")

    return record

def _infer_preferred_stance_from_statement(sents: list[str], committee_stance: Optional[str]) -> Optional[str]:
    """
    Secondary preferred-stance inference when no explicit label was found in the
    vote sentence itself.  Called after _extract_member_stance_vote_from_statement
    returns None for preferred.

    Pass 1 -- explicit "should be [stance]" anywhere in the statement.
              Handles Dec2023: "If at all there is a stance, it should be neutral."

    Pass 2 -- "pause" ONLY inside sentences that also contain a stance/accommodation/
              tightening keyword, so rate-pause references (Dec2022: "argued for a
              pause") are excluded.
              Handles Oct2022: "pause rather than focus on further tightening."
              Does NOT fire for Dec2022 (pause is in a rate-only sentence).
    """
    # Pass 1: explicit preference via "should be [stance]"
    for sent in sents:
        m = re.search(
            r'\bshould\s+be\s+(neutral|accommodative|calibrated\s+tightening)\b',
            sent, re.IGNORECASE
        )
        if m:
            return _normalise_stance(m.group(1))

    # Pass 2: "pause" in stance-context sentences only
    if committee_stance == "withdrawal of accommodation":
        stance_context = " ".join(
            s for s in sents
            if re.search(r'\bstance\b|\baccommodation\b|\btightening\b', s, re.IGNORECASE)
        )
        if re.search(r'\bpause\b', stance_context, re.IGNORECASE):
            return "neutral"
    # Pass 3: positive preference via "needs to be / ought to be / prefer [stance]"
    for sent in sents:
        m = re.search(
            r'\b(?:stance\s+(?:needs?\s+to|ought\s+to|should)\s+be|'
            r'preference\s+(?:is\s+)?(?:for\s+)?(?:a\s+)?|'
            r'prefer(?:ring)?\s+(?:a\s+)?(?:neutral|accommodative|calibrated\s+tightening)\s+stance)'
            r'(?:\s+(?:neutral|accommodative|calibrated\s+tightening))?',
            sent, re.IGNORECASE
        )
        if m:
            label = _stance_label_in_text(sent)
            if label:
                return label
            
    # Pass 4: "move/shift/transition/adopt/change to a [stance] stance"
    # Handles Goyal Dec2022: "It is time to move to a neutral stance"
    for sent in sents:
        m = re.search(
            r'\b(?:move|shift|transition|adopt|change)\s+to\s+(?:a\s+)?'
            r'(neutral|accommodative|calibrated\s+tightening)\s+stance\b',
            sent, re.IGNORECASE
        )
        if m:
            return _normalise_stance(m.group(1))

    return None

def _semantic_validation_flags(full_text: str, rec: dict) -> dict[str, int]:
    members = rec.get("members", [])
    pre_statement_text = re.split(r'\bStatement\s+by\b', full_text, maxsplit=1, flags=re.IGNORECASE)[0]
    pre_norm = _normalize_for_regex(pre_statement_text)
    stance_sources = rec.get("stance_vote_sources", {})
    stance_votes = rec.get("stance_votes", {})
    flags = {
        "flag_explicit_mpc_stance_statement_ignored": 0,
        "flag_explicit_member_vote_overwritten_by_inference": 0,
        "flag_stance_action_inconsistent_with_change_language": 0,
        "flag_missing_delta_despite_status_quo": 0,
        "flag_unanimous_stance_statement_missing_votes": 0,
        "flag_member_statement_explicit_overwritten_by_inference": 0,
    }

    has_mpc_stance_evidence = bool(re.search(
        r'(?:voted|unanimously|expressed\s+reservations?|were\s+of\s+the\s+view|did\s+not\s+(?:agree|support)).{0,260}'
        r'(?:stance|withdrawal\s+of\s+accommodation)',
        pre_norm, re.IGNORECASE
    ))
    has_mpc_source = any(str(v).startswith("mpc_wide") or str(v) == "member_unanimous_mpc_statement"
                         for v in stance_sources.values())
    all_high_quality = all(
        str(v) in ("member_statement_explicit", "missing_dissent_stance")
        for v in stance_sources.values()
    )
    if has_mpc_stance_evidence and stance_votes and not has_mpc_source and not all_high_quality:
        flags["flag_explicit_mpc_stance_statement_ignored"] = 1

    if re.search(r'\b(?:MPC|committee)\b.{0,80}\bdecided\s+to\s+change\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance\b|voted\s+in\s+favou?r\s+of\s+changing\s+(?:the\s+)?(?:monetary\s+policy\s+)?stance', pre_norm, re.IGNORECASE):
        if rec.get("res_pol_stance_action") != "change" and not re.search(r'\b(?:not|against|reservations?)\b.{0,80}\bchange', pre_norm, re.IGNORECASE):
            flags["flag_stance_action_inconsistent_with_change_language"] = 1

    if re.search(r'\b(?:unanimously|all\s+members)\b.{0,260}(?:stance|withdrawal\s+of\s+accommodation)', pre_norm, re.IGNORECASE):
        if len(stance_votes) < len(members):
            flags["flag_unanimous_stance_statement_missing_votes"] = 1

    inferred_map = rec.get("stance_vote_member_inferred_map", {})
    for mem in members:
        name = mem.get("name", "")
        if _find_vote(name, inferred_map):
            stmt = _find_statement(name, rec.get("statements", {}))
            explicit, _preferred = _extract_member_stance_vote_from_statement(stmt, rec.get("res_pol_stance")) if stmt else (None, None)
            if explicit:
                flags["flag_explicit_member_vote_overwritten_by_inference"] = 1
                flags["flag_member_statement_explicit_overwritten_by_inference"] = 1

        delta = _find_vote(name, rec.get("member_preferred_deltas", {}))
        rate_v = _find_vote(name, rec.get("rate_votes", {}))
        if delta in ("", None) and rate_v != "Yes":
            for sent in _sentences(pre_statement_text + " " + _find_statement(name, rec.get("statements", {}))):
                if _member_names_in_text(sent, [mem]) and re.search(r'\b(?:status\s+quo|unchanged|no\s+change)\b', sent, re.IGNORECASE):
                    flags["flag_missing_delta_despite_status_quo"] = 1
                    break
    return flags

# ── Exporters ──────────────────────────────────────────────────────────────────

def build_meeting_level(records: list[dict]) -> pd.DataFrame:
    rows = []
    for rec in records:
        row = {
            "file_name":              rec.get("file_name", ""),
            "start_date":             _fmt_date(rec.get("start_date")),
            "end_date":               _fmt_date(rec.get("end_date")),
            "res_pol_rate_repo":      rec.get("res_pol_rate_repo"),
            "res_pol_rate_action":    rec.get("res_pol_rate_action", ""),
            "res_pol_stance":         rec.get("res_pol_stance", ""),
            "res_pol_stance_action":  rec.get("res_pol_stance_action", ""),
            "matched_rate_sentence":  rec.get("matched_rate_sentence", ""),
            "matched_stance_sentence":rec.get("matched_stance_sentence", ""),
            "stance_vote_unanimous_inferred": bool(rec.get("stance_vote_unanimous_inferred", False)),
            "stance_vote_member_inferred": bool(rec.get("stance_vote_member_inferred", False)),
        }
        members      = rec.get("members", [])
        rate_votes   = rec.get("rate_votes", {})
        stance_votes = rec.get("stance_votes", {})
        member_inferred = rec.get("stance_vote_member_inferred_map", {})
        preferred_stance = rec.get("member_preferred_stance", {})
        statements   = rec.get("statements", {})

        for slot in range(1, MAX_MEMBERS + 1):
            if slot <= len(members):
                mem  = members[slot - 1]
                name = mem.get("name", "")
                sv   = _find_vote(name, stance_votes)
                row[f"mem_{slot}"]              = name
                row[f"designation_mem_{slot}"]  = mem.get("designation", "")
                row[f"rate_vote_mem_{slot}"]    = _find_vote(name, rate_votes)
                row[f"stance_vote_mem_{slot}"]  = sv
                row[f"stance_vote_member_inferred_mem_{slot}"] = bool(_find_vote(name, member_inferred))
                row[f"preferred_stance_mem_{slot}"] = _find_vote(name, preferred_stance)
                row[f"stmt_mem_{slot}"]         = _find_statement(name, statements)
            else:
                for suf in ["", "_designation", "_rate_vote", "_stance_vote", "_stmt"]:
                    row[f"mem_{slot}{suf.replace('_','')}"] = ""
                row[f"mem_{slot}"] = row[f"designation_mem_{slot}"] = ""
                row[f"rate_vote_mem_{slot}"] = row[f"stance_vote_mem_{slot}"] = row[f"stmt_mem_{slot}"] = ""
                row[f"stance_vote_member_inferred_mem_{slot}"] = False
                row[f"preferred_stance_mem_{slot}"] = ""
        rows.append(row)
    return pd.DataFrame(rows)


def build_member_level_long(records: list[dict]) -> pd.DataFrame:
    rows = []
    prev_by_file = _previous_repo_by_file(records)
    for rec in records:
        rate_votes   = rec.get("rate_votes", {})
        stance_votes = rec.get("stance_votes", {})
        member_inferred = rec.get("stance_vote_member_inferred_map", {})
        stance_sources = rec.get("stance_vote_sources", {})
        preferred_stance = rec.get("member_preferred_stance", {})
        statements   = rec.get("statements", {})
        prev_rate = prev_by_file.get(rec.get("file_name", ""))
        for slot, mem in enumerate(rec.get("members", []), 1):
            name = mem.get("name", "")
            sv   = _find_vote(name, stance_votes)
            rate_fields = _member_preferred_rate_fields(name, rec, prev_rate)
            row = {
                "file_name":     rec.get("file_name", ""),
                "start_date":    _fmt_date(rec.get("start_date")),
                "end_date":      _fmt_date(rec.get("end_date")),
                "member_slot":   f"mem_{slot}",
                "member_name":   name,
                "designation":   mem.get("designation", ""),
                "is_governor":   mem.get("is_governor", False),
                "rate_vote":     _find_vote(name, rate_votes),
                "stance_vote":   sv,
                "stance_vote_inferred": bool(rec.get("stance_vote_unanimous_inferred", False) or _find_vote(name, member_inferred)),
                "stance_vote_unanimous_inferred": bool(rec.get("stance_vote_unanimous_inferred", False)),
                "stance_vote_member_inferred": bool(_find_vote(name, member_inferred)),
                "stance_vote_source": _find_vote(name, stance_sources),
                "member_preferred_stance": _find_vote(name, preferred_stance),
                "statement":     _find_statement(name, statements),
                "res_pol_rate_repo":   rec.get("res_pol_rate_repo"),
                "res_pol_rate_action": rec.get("res_pol_rate_action", ""),
                "res_pol_stance":      rec.get("res_pol_stance", ""),
                "res_pol_stance_action":rec.get("res_pol_stance_action", ""),
            }
            row.update(rate_fields)
            rows.append(row)
    return pd.DataFrame(rows)


def build_extraction_log(records: list[dict]) -> pd.DataFrame:
    rows = []
    for rec in records:
        members    = rec.get("members", [])
        rate_votes = rec.get("rate_votes", {})
        statements = rec.get("statements", {})
        sv_dict    = rec.get("stance_votes", {})
        inferred   = rec.get("stance_vote_inferred", False)
        member_inferred = rec.get("stance_vote_member_inferred_map", {})
        semantic_flags = rec.get("semantic_validation_flags", {})
        rows.append({
            "file_name":            rec.get("file_name", ""),
            "start_date":           _fmt_date(rec.get("start_date")),
            "end_date":             _fmt_date(rec.get("end_date")),
            "members_found":        len(members),
            "votes_found":          len(rate_votes),
            "statements_found":     len(statements),
            "stance_votes_found":   len(sv_dict),
            "stance_vote_inferred": int(inferred),
            "stance_vote_unanimous_inferred": int(rec.get("stance_vote_unanimous_inferred", False)),
            "stance_vote_member_inferred": int(rec.get("stance_vote_member_inferred", False)),
            "stance_votes_member_inferred_count": len(member_inferred),
            "date_rule":            rec.get("matched_rule", ""),
            "date_candidate":       rec.get("candidate_text", ""),
            "flag_date_failed":     int(not rec.get("start_date")),
            "flag_missing_rate":    int(rec.get("res_pol_rate_repo") is None),
            "flag_missing_stance":  int(not rec.get("res_pol_stance")),
            "flag_less_than_6_members": int(len(members) < 6),
            "flag_missing_votes":   int(len(rate_votes) < len(members)),
            "flag_missing_statements": int(len(statements) < len(members)),
            "flag_missing_stance_votes": int(len(sv_dict) == 0),
            "flag_incomplete_stance_votes": int(0 < len(sv_dict) < len(members)),
            "flag_stance_from_fallback": int(rec.get("stance_from_fallback", False)),
            "flag_explicit_mpc_stance_statement_ignored": int(semantic_flags.get("flag_explicit_mpc_stance_statement_ignored", 0)),
            "flag_explicit_member_vote_overwritten_by_inference": int(semantic_flags.get("flag_explicit_member_vote_overwritten_by_inference", 0)),
            "flag_stance_action_inconsistent_with_change_language": int(semantic_flags.get("flag_stance_action_inconsistent_with_change_language", 0)),
            "flag_missing_delta_despite_status_quo": int(semantic_flags.get("flag_missing_delta_despite_status_quo", 0)),
            "flag_unanimous_stance_statement_missing_votes": int(semantic_flags.get("flag_unanimous_stance_statement_missing_votes", 0)),
            "flag_member_statement_explicit_overwritten_by_inference": int(semantic_flags.get("flag_member_statement_explicit_overwritten_by_inference", 0)),
            "issues": "; ".join(rec.get("issues_summary", [])),
        })
    return pd.DataFrame(rows)


def _format_sheet(ws) -> None:
    header_fill = PatternFill(start_color="1F497D", end_color="1F497D", fill_type="solid")
    header_font = Font(color="FFFFFF", bold=True)
    for cell in ws[1]:
        cell.fill   = header_fill
        cell.font   = header_font
        cell.alignment = Alignment(horizontal="center", wrap_text=True)
    for col in ws.columns:
        max_len = max((len(str(c.value)) for c in col if c.value), default=0)
        ws.column_dimensions[col[0].column_letter].width = min(max_len + 2, 60)
    ws.freeze_panes = "A2"


def export_to_excel(records: list[dict], output_path: str) -> None:
    df_meeting = build_meeting_level(records)
    df_long    = build_member_level_long(records)
    df_log     = build_extraction_log(records)

    logger.info(f"Writing output to {output_path}")
    with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
        df_meeting.to_excel(writer, sheet_name="meeting_level",    index=False)
        df_long.to_excel(   writer, sheet_name="member_level_long",index=False)
        df_log.to_excel(    writer, sheet_name="extraction_log",   index=False)
        wb = writer.book
        for sheet_name in writer.sheets:
            _format_sheet(wb[sheet_name])

    logger.info(f"Export complete: {len(records)} meetings, {len(df_long)} member rows")


# ── Cross-Meeting Semantic Validator ───────────────────────────────────────────

def validate_cross_meeting(records: list[dict]) -> list[str]:
    """Run post-extraction consistency checks across all meetings.
    
    Returns a list of human-readable warning strings.
    Does NOT modify records — this is a validation layer, not a correction layer.
    """
    warnings = []
    
    # Sort by start_date chronologically
    sorted_recs = sorted(records, key=lambda r: r.get("start_date") or date(1900,1,1))
    
    prev = None
    for rec in sorted_recs:
        fname    = rec.get("file_name", "?")
        rate     = rec.get("res_pol_rate_repo")
        action   = rec.get("res_pol_rate_action")
        members  = rec.get("members", [])
        rv       = rec.get("rate_votes", {})
        
        # 1. Rate consistency: hikes increase, cuts decrease, unchanged stays same
        if prev and rate is not None and prev.get("res_pol_rate_repo") is not None:
            prev_rate = prev["res_pol_rate_repo"]
            if abs(rate - prev_rate) > 2.0:
                msg = f"IMPOSSIBLE_RATE_JUMP: {fname} repo rate jump {prev_rate} -> {rate}"
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
            if action == "unchanged" and abs(rate - prev_rate) > 0.01:
                msg = (f"RATE_CONSISTENCY: {fname} action=unchanged but rate "
                       f"{rate} != prev {prev_rate} ({prev['file_name']})")
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
            elif action == "cut" and rate >= prev_rate:
                msg = (f"RATE_CONSISTENCY: {fname} action=cut but rate "
                       f"{rate} >= prev {prev_rate} ({prev['file_name']})")
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
            elif action == "hike" and rate <= prev_rate:
                msg = (f"RATE_CONSISTENCY: {fname} action=hike but rate "
                       f"{rate} <= prev {prev_rate} ({prev['file_name']})")
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
        
        # 2. Member validation
        if len(members) != 6:
            msg = f"MEMBER_COUNT: {fname} has {len(members)} members (expected 6)"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)
        
        gov_count = sum(1 for m in members if m.get("is_governor"))
        if gov_count != 1:
            msg = f"GOVERNOR_COUNT: {fname} has {gov_count} governors (expected 1)"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)
        
        names = [m.get("name","") for m in members]
        canonical_names = [_canonical_name(n) for n in names]
        if len(canonical_names) != len(set(canonical_names)):
            msg = f"DUPLICATE_MEMBER: {fname} has duplicate member names: {names}"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)
        
        # 3. Vote count vs member count
        if len(rv) < len(members):
            msg = f"VOTE_COVERAGE: {fname} has {len(rv)} rate votes for {len(members)} members"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)
        if len(rec.get("stance_votes", {})) not in (0, len(members)):
            msg = f"STANCE_COVERAGE: {fname} has {len(rec.get('stance_votes', {}))} stance votes for {len(members)} members"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)
        
        # 4. Temporal: dates should be strictly increasing
        if prev and rec.get("start_date") and prev.get("start_date"):
            if rec["start_date"] <= prev["start_date"]:
                msg = (f"CHRONOLOGY: {fname} start_date {rec['start_date']} "
                       f"<= prev {prev['start_date']} ({prev['file_name']})")
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
        
        # 5. Stance from fallback (higher risk of contamination)
        if rec.get("stance_from_fallback"):
            msg = f"STANCE_FALLBACK: {fname} stance extracted from full-text fallback, not Resolution section"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)

        matched_stance_sentence = rec.get("matched_stance_sentence") or ""
        if matched_stance_sentence and not _has_stance_phrase(matched_stance_sentence):
            msg = f"STANCE_SENTENCE_INVALID: {fname} matched stance sentence lacks a valid stance phrase"
            warnings.append(msg)
            rec.setdefault("issues_summary", []).append(msg)

        for name, stmt in rec.get("statements", {}).items():
            if re.search(r'\bStatement\s+by\b', stmt, re.IGNORECASE):
                msg = f"STATEMENT_BLEED: {fname} statement for {name} contains another statement heading"
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)

        for flag_name, value in rec.get("semantic_validation_flags", {}).items():
            if value:
                msg = f"{flag_name.upper()}: {fname}"
                warnings.append(msg)
                rec.setdefault("issues_summary", []).append(msg)
        
        prev = rec
    
    return warnings

# ── Runner ─────────────────────────────────────────────────────────────────────

if __name__ == "__main__":
    logger.info("=" * 70)
    logger.info("RBI MPC Minutes Extraction Pipeline v2")
    logger.info("=" * 70)
    logger.info(f"PDF directory : {PDF_DIR}")
    logger.info(f"Output path   : {OUTPUT_PATH}")

    if not os.path.isdir(PDF_DIR):
        raise FileNotFoundError(f"PDF directory not found: {PDF_DIR}")

    pdf_paths = sorted(str(p) for p in Path(PDF_DIR).glob("*.pdf"))
    logger.info(f"Found {len(pdf_paths)} PDFs")

    records = []
    for i, path in enumerate(pdf_paths, 1):
        fname = os.path.basename(path)
        logger.info(f"\n[{i:02d}/{len(pdf_paths)}] {fname}")
        t0 = time.time()
        record = process_pdf(path)
        elapsed = time.time() - t0
        records.append(record)
        logger.info(
            f"  done {fname} | {elapsed:.1f}s | "
            f"members={len(record.get('members', []))} "
            f"stmts={len(record.get('statements', {}))} "
            f"rate_votes={len(record.get('rate_votes', {}))} "
            f"stance_votes={len(record.get('stance_votes', {}))} "
            f"(inferred={record.get('stance_vote_inferred', False)})"
        )

    logger.info("\n" + "=" * 70)
    logger.info(f"SUMMARY: {len(records)} PDFs processed")

    # F1: Cross-meeting stance_action correction.
    # `_determine_stance_action` runs per-meeting and cannot tell whether
    # "decided to remain focused on withdrawal of accommodation" is a continuation
    # (Aug2022 onwards) or a change (June2022, the first such meeting). Correct
    # by comparing each meeting's stance label to the immediately prior meeting.
    from datetime import datetime, date
    def _sort_key(r):
        sd = r.get("start_date")
        if sd is None:
            return datetime.max
        if isinstance(sd, date):  # date or datetime
            return datetime(sd.year, sd.month, sd.day) if not isinstance(sd, datetime) else sd
        if isinstance(sd, str):
            try:
                return datetime.strptime(sd, "%d/%m/%Y")
            except (ValueError, TypeError):
                return datetime.max
        return datetime.max
    sorted_records = sorted(records, key=_sort_key)
    prior_stance = None
    for rec in sorted_records:
        curr_stance = rec.get("res_pol_stance")
        if curr_stance and prior_stance and curr_stance != prior_stance:
            if rec.get("res_pol_stance_action") == "continue":
                rec["res_pol_stance_action"] = "change"
                logger.info(
                    f"[{rec.get('file_name', '?')}] stance_action corrected: "
                    f"continue -> change (prior={prior_stance}, curr={curr_stance})"
                )
        if curr_stance:
            prior_stance = curr_stance

    # Cross-meeting validation
    validation_warnings = validate_cross_meeting(records)
    if validation_warnings:
        logger.info(f"\n--- CROSS-MEETING VALIDATION: {len(validation_warnings)} warnings ---")
        for w in validation_warnings:
            logger.warning(f"  {w}")
    else:
        logger.info("Cross-meeting validation: ALL CHECKS PASSED")
    
    export_to_excel(records, OUTPUT_PATH)
    logger.info("Pipeline finished.")
```

```python
#!/usr/bin/env python3
"""
Cross-check an MPC minutes xlsx file against a manually-verified ground truth
extracted from the RBI press releases (Dec 2016 - Apr 2026, 58 meetings).

Usage:
    python3 crosscheck.py path/to/mpc_minutes_extracted.xlsx

Expected xlsx structure:
    Sheet 'meeting_level' with cols: file_name, start_date, end_date,
        res_pol_rate_repo, res_pol_rate_action, res_pol_stance,
        res_pol_stance_action, ...
    Sheet 'member_level_long' with cols: file_name, start_date, end_date,
        member_slot, member_name, designation, is_governor,
        rate_vote, stance_vote, ..., member_preferred_stance, ...,
        member_preferred_delta_bps, ...

The start_date column is expected to be a string in dd/mm/yyyy format.

Methodological conventions used in the ground truth (must match xlsx
encoding for the script to pass):
1. Magnitude-split rate dissent: when the committee voted for one magnitude
   but a member preferred a different magnitude (Aug 2019, Oct 2019,
   Mar 2020, May 2020, Jun 2025), encode rate_vote as "No" for the
   quantum-dissenters even though the literal RBI voting table records
   "Yes". This treats any magnitude deviation as a dissent for downstream
   consistency.
2. Oct 2020 Varma: he voted Yes on the accommodative stance; his
   objection was to the time-horizon formulation only - treated as
   stance_vote = Yes, preferred = accommodative.

Ground-truth format:
    GT[start_date_ddmmyyyy] = {
        "rate":         committee final rate (float),
        "rate_action":  "hike" / "cut" / "unchanged",
        "stance":       committee stance,
        "prev_rate":    prior meeting's rate,
        "members": {
            "Full Name": {
                "rate_vote":         "Yes" / "No",
                "stance_vote":       "Yes" / "No",
                "preferred_stance":  str or None,
                "delta_bps":         int (member's preferred change from prev_rate)
            },
            ...
        }
    }
"""

XLSX_PATH = r""
import openpyxl

GT = {}

# Dec 6-7, 2016 (06/12/2016) - 6.25% unchanged, Accom, unanimous
GT["06/12/2016"] = {
    "rate": 6.25, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri R. Gandhi":               {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Feb 7-8, 2017 (07/02/2017) - 6.25% unchanged, stance changed to Neutral, 6-0
GT["07/02/2017"] = {
    "rate": 6.25, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Apr 5-6, 2017 (05/04/2017) - 6.25% unchanged, Neutral, 6-0
GT["05/04/2017"] = {
    "rate": 6.25, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Jun 6-7, 2017 (06/06/2017) - 6.25% unchanged, Neutral, 5-1 (Dholakia preferred -50)
GT["06/06/2017"] = {
    "rate": 6.25, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Aug 1-2, 2017 (01/08/2017) - Cut -25 to 6.00%, Neutral, 4-2 (Dholakia -50; Patra status quo)
GT["01/08/2017"] = {
    "rate": 6.0, "rate_action": "cut", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": -50},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
    }
}

# Oct 3-4, 2017 (03/10/2017) - 6.00% unchanged, Neutral, 5-1 (Dholakia at least -25)
GT["03/10/2017"] = {
    "rate": 6.0, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},  # "at least -25"
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Dec 5-6, 2017 (05/12/2017) - 6.00% unchanged, Neutral, 5-1 (Dholakia -25)
GT["05/12/2017"] = {
    "rate": 6.0, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Feb 6-7, 2018 (06/02/2018) - 6.00% unchanged, Neutral, 5-1 (Patra +25)
GT["06/02/2018"] = {
    "rate": 6.0, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "withdrawal of accommodation", "delta_bps": 25},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Apr 4-5, 2018 (04/04/2018) - 6.00% unchanged, Neutral, 5-1 (Patra +25)
GT["04/04/2018"] = {
    "rate": 6.0, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": None, "delta_bps": 25},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Jun 4-6, 2018 (04/06/2018) - Hike +25 to 6.25%, Neutral, 6-0
GT["04/06/2018"] = {
    "rate": 6.25, "rate_action": "hike", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
    }
}

# Jul 30 - Aug 1, 2018 (30/07/2018) - Hike +25 to 6.50%, Neutral, 5-1 (Dholakia against, wanted status quo)
GT["30/07/2018"] = {
    "rate": 6.5, "rate_action": "hike", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 25},
    }
}

# Oct 3-5, 2018 (03/10/2018) - 6.50% unchanged, change stance to Calibrated Tightening, rate 5-1 (Ghate +25), stance 5-1 (Dholakia Neutral)
GT["03/10/2018"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "calibrated tightening", "prev_rate": 6.5,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
    }
}

# Dec 3-5, 2018 (03/12/2018) - 6.50% unchanged, maintain Calibrated Tightening, rate 6-0, stance 5-1 (Dholakia Neutral)
GT["03/12/2018"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "calibrated tightening", "prev_rate": 6.5,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
        "Dr. Urjit R. Patel":           {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "calibrated tightening", "delta_bps": 0},
    }
}

# Feb 5-7, 2019 (05/02/2019) - Cut -25 to 6.25%, change stance to Neutral, rate 4-2 (Ghate, Acharya unchanged), stance 6-0
GT["05/02/2019"] = {
    "rate": 6.25, "rate_action": "cut", "stance": "neutral", "prev_rate": 6.5,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Viral V. Acharya":         {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
    }
}

# Apr 2-4, 2019 (02/04/2019) - Cut -25 to 6.00%, Neutral, rate 4-2 (Ghate, Acharya unchanged), stance 5-1 (Dholakia Accom)
GT["02/04/2019"] = {
    "rate": 6.0, "rate_action": "cut", "stance": "neutral", "prev_rate": 6.25,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Viral V. Acharya":         {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
    }
}

# Jun 3,4,6, 2019 (03/06/2019) - Cut -25 to 5.75%, change stance to Accom, both unanimous 6-0
GT["03/06/2019"] = {
    "rate": 5.75, "rate_action": "cut", "stance": "accommodative", "prev_rate": 6.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Viral V. Acharya":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
    }
}

# Aug 5-7, 2019 (05/08/2019) - Cut -35 to 5.40%, Accom, rate all "Yes" but quantum split (Ghate, Dua -25; rest -35)
GT["05/08/2019"] = {
    "rate": 5.4, "rate_action": "cut", "stance": "accommodative", "prev_rate": 5.75,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Pami Dua":                 {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -35},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -35},
        "Shri Bibhu Prasad Kanungo":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -35},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -35},
    }
}

# Oct 1,3,4, 2019 (01/10/2019) - Cut -25 to 5.15%, Accom, all "Yes" rate; Dholakia -40 (others -25)
GT["01/10/2019"] = {
    "rate": 5.15, "rate_action": "cut", "stance": "accommodative", "prev_rate": 5.4,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Bibhu Prasad Kanungo":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
    }
}

# Dec 3-5, 2019 (03/12/2019) - 5.15% unchanged, Accom, both unanimous
GT["03/12/2019"] = {
    "rate": 5.15, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 5.15,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Bibhu Prasad Kanungo":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Feb 4-6, 2020 (04/02/2020) - 5.15% unchanged, Accom, both unanimous
GT["04/02/2020"] = {
    "rate": 5.15, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 5.15,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Janak Raj":                {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Mar 24,26,27, 2020 (24/03/2020) - Cut -75 to 4.40%, Accom, all Yes; Ghate, Dua -50; rest -75
GT["24/03/2020"] = {
    "rate": 4.4, "rate_action": "cut", "stance": "accommodative", "prev_rate": 5.15,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -50},
        "Dr. Pami Dua":                 {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -50},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -75},
        "Dr. Janak Raj":                {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -75},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -75},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -75},
    }
}

# May 20-22, 2020 (20/05/2020) - Cut -40 to 4.00%, Accom, all Yes; Ghate -25, rest -40
GT["20/05/2020"] = {
    "rate": 4.0, "rate_action": "cut", "stance": "accommodative", "prev_rate": 4.4,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
        "Dr. Janak Raj":                {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -40},
    }
}

# Aug 4-6, 2020 (04/08/2020) - 4.00% unchanged, Accom, both unanimous
GT["04/08/2020"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Chetan Ghate":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Pami Dua":                 {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ravindra H. Dholakia":     {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Oct 7-9, 2020 (07/10/2020) - 4.00% unchanged, Accom; Varma voted against time-horizon formulation (treat as Yes Accom but No on formulation)
# Per resolution: "All members ... unanimously voted for keeping the policy repo rate unchanged and continue with the accommodative stance...
# Dr. Shashanka Bhide, Dr. Ashima Goyal, Dr. Mridul K. Saggar, Dr. Michael Debabrata Patra and Shri Shaktikanta Das voted to continue
# with this accommodative stance at least during the current financial year and into the next financial year, with Prof. Jayanth R. Varma
# voting against this formulation."
# I'll mark Varma stance_vote as "No" on this specific formulation. Preferred_stance is still accommodative.
GT["07/10/2020"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Dec 2-4, 2020 (02/12/2020) - 4.00% unchanged, Accom, both unanimous
GT["02/12/2020"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Feb 3-5, 2021 (03/02/2021) - 4.00% unchanged, Accom, both unanimous
GT["03/02/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Apr 5-7, 2021 (05/04/2021) - 4.00% unchanged, Accom, both unanimous
GT["05/04/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Jun 2-4, 2021 (02/06/2021) - 4.00% unchanged, Accom, both unanimous
GT["02/06/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Aug 4-6, 2021 (04/08/2021) - 4.00% unchanged, Accom; rate 6-0; Varma stance reservations (No)
GT["04/08/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Oct 6-8, 2021 (06/10/2021) - 4.00% unchanged, Accom; rate 6-0; Varma stance No
GT["06/10/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Dec 6-8, 2021 (06/12/2021) - 4.00% unchanged, Accom; rate 6-0; Varma stance No
GT["06/12/2021"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Feb 8-10, 2022 (08/02/2022) - 4.00% unchanged, Accom; rate 6-0; Varma stance No
GT["08/02/2022"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# Apr 6-8, 2022 (06/04/2022) - 4.00% unchanged, "Accommodative while focusing on withdrawal of accommodation", both unanimous
GT["06/04/2022"] = {
    "rate": 4.0, "rate_action": "unchanged", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Mridul K. Saggar":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 0},
    }
}

# May 2,4, 2022 (02/05/2022) - Hike +40 to 4.40%, "Accommodative w/ focus on withdrawal", both unanimous
GT["02/05/2022"] = {
    "rate": 4.4, "rate_action": "hike", "stance": "accommodative", "prev_rate": 4.0,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": 40},
    }
}

# Jun 6-8, 2022 (06/06/2022) - Hike +50 to 4.90%, Withdrawal of Accommodation, both unanimous
GT["06/06/2022"] = {
    "rate": 4.9, "rate_action": "hike", "stance": "withdrawal of accommodation", "prev_rate": 4.4,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
    }
}

# Aug 3-5, 2022 (03/08/2022) - Hike +50 to 5.40%, WoA; rate 6-0; Varma stance No (reservations)
GT["03/08/2022"] = {
    "rate": 5.4, "rate_action": "hike", "stance": "withdrawal of accommodation", "prev_rate": 4.9,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 50},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
    }
}

# Sep 28-30, 2022 (28/09/2022) - Hike +50 to 5.90%, WoA; rate 5-1 (Goyal No, wanted +35); stance 5-1 (Varma No)
GT["28/09/2022"] = {
    "rate": 5.9, "rate_action": "hike", "stance": "withdrawal of accommodation", "prev_rate": 5.4,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Ashima Goyal":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 35},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 50},  # Varma against WoA — he wants neutral
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 50},
    }
}

# Dec 5-7, 2022 (05/12/2022) - Hike +35 to 6.25%, WoA; rate 5-1 (Varma No); stance 4-2 (Goyal, Varma No)
# Varma wanted -??-: He voted against the hike. What did he prefer? His statement is needed.
# For now: rate dissent = stop hiking (delta = 0); stance dissent = neutral
GT["05/12/2022"] = {
    "rate": 6.25, "rate_action": "hike", "stance": "withdrawal of accommodation", "prev_rate": 5.9,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 35},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 35},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},  # against hike
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 35},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 35},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 35},
    }
}

# Feb 6-8, 2023 (06/02/2023) - Hike +25 to 6.50%, WoA; rate 4-2 (Goyal, Varma No); stance 4-2 (Goyal, Varma No)
GT["06/02/2023"] = {
    "rate": 6.5, "rate_action": "hike", "stance": "withdrawal of accommodation", "prev_rate": 6.25,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 25},
        "Dr. Ashima Goyal":             {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 25},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 25},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 25},
    }
}

# Apr 3,5,6, 2023 (03/04/2023) - 6.50% unchanged, WoA; rate 6-0; Varma stance No (reservations)
GT["03/04/2023"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Jun 6-8, 2023 (06/06/2023) - 6.50% unchanged, WoA; rate 6-0; Varma stance No
GT["06/06/2023"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Aug 8-10, 2023 (08/08/2023) - 6.50% unchanged, WoA; rate 6-0; Varma stance No
GT["08/08/2023"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Oct 4-6, 2023 (04/10/2023) - 6.50% unchanged, WoA; rate 6-0; Varma stance No
GT["04/10/2023"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": None, "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Dec 6-8, 2023 (06/12/2023) - 6.50% unchanged, WoA; rate 6-0; Varma stance No
GT["06/12/2023"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Feb 6-8, 2024 (06/02/2024) - 6.50% unchanged, WoA; rate 5-1 (Varma -25); stance 5-1 (Varma Neutral)
GT["06/02/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Apr 3-5, 2024 (03/04/2024) - 6.50% unchanged, WoA; rate 5-1; stance 5-1
GT["03/04/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Jun 5-7, 2024 (05/06/2024) - 6.50% unchanged, WoA; rate 4-2 (Goyal, Varma No); stance 4-2 (Goyal, Varma Neutral)
GT["05/06/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Aug 6-8, 2024 (06/08/2024) - 6.50% unchanged, WoA; rate 4-2; stance 4-2
GT["06/08/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "withdrawal of accommodation", "prev_rate": 6.5,
    "members": {
        "Dr. Shashanka Bhide":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Ashima Goyal":             {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Prof. Jayanth R. Varma":       {"rate_vote": "No",  "stance_vote": "No",  "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "withdrawal of accommodation", "delta_bps": 0},
    }
}

# Oct 7-9, 2024 (07/10/2024) - 6.50% unchanged, change stance to Neutral; rate 5-1 (Kumar -25); stance 6-0
GT["07/10/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.5,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Dec 4-6, 2024 (04/12/2024) - 6.50% unchanged, Neutral; rate 4-2 (Kumar, Singh -25); stance 6-0
GT["04/12/2024"] = {
    "rate": 6.5, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 6.5,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "No",  "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Michael Debabrata Patra":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Shaktikanta Das":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Feb 5-7, 2025 (05/02/2025) - Cut -25 to 6.25%, Neutral; both unanimous 6-0
GT["05/02/2025"] = {
    "rate": 6.25, "rate_action": "cut", "stance": "neutral", "prev_rate": 6.5,
    "members": {
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri M Rajeshwar Rao":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
    }
}

# Apr 7-9, 2025 (07/04/2025) - Cut -25 to 6.00%, change to Accommodative; both unanimous
GT["07/04/2025"] = {
    "rate": 6.0, "rate_action": "cut", "stance": "accommodative", "prev_rate": 6.25,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri M Rajeshwar Rao":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "accommodative", "delta_bps": -25},
    }
}

# Jun 4-6, 2025 (04/06/2025) - Cut -50 to 5.50%, change to Neutral; rate all Yes but quantum split (Bhattacharya -25); stance 6-0
GT["04/06/2025"] = {
    "rate": 5.5, "rate_action": "cut", "stance": "neutral", "prev_rate": 6.0,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
        "Shri Saugata Bhattacharya":    {"rate_vote": "No", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -50},
    }
}

# Aug 4-6, 2025 (04/08/2025) - 5.50% unchanged, Neutral; both unanimous
GT["04/08/2025"] = {
    "rate": 5.5, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 5.5,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Rajiv Ranjan":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Sep 29-Oct 1, 2025 (29/09/2025) - 5.50% unchanged, Neutral; rate 6-0; stance 4-2 (Kumar, Singh wanted Accom)
GT["29/09/2025"] = {
    "rate": 5.5, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 5.5,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Indranil Bhattacharyya":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Dec 3-5, 2025 (03/12/2025) - Cut -25 to 5.25%, Neutral; rate 6-0; stance 5-1 (Singh Accom)
GT["03/12/2025"] = {
    "rate": 5.25, "rate_action": "cut", "stance": "neutral", "prev_rate": 5.5,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": -25},
        "Shri Indranil Bhattacharyya":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": -25},
    }
}

# Feb 4-6, 2026 (04/02/2026) - 5.25% unchanged, Neutral; rate 6-0; stance 5-1 (Singh Accom)
GT["04/02/2026"] = {
    "rate": 5.25, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 5.25,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "No",  "preferred_stance": "accommodative", "delta_bps": 0},
        "Shri Indranil Bhattacharyya":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}

# Apr 6-8, 2026 (06/04/2026) - 5.25% unchanged, Neutral; both unanimous 6-0
GT["06/04/2026"] = {
    "rate": 5.25, "rate_action": "unchanged", "stance": "neutral", "prev_rate": 5.25,
    "members": {
        "Dr. Nagesh Kumar":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Saugata Bhattacharya":    {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Professor Ram Singh":          {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Indranil Bhattacharyya":  {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Dr. Poonam Gupta":             {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
        "Shri Sanjay Malhotra":         {"rate_vote": "Yes", "stance_vote": "Yes", "preferred_stance": "neutral", "delta_bps": 0},
    }
}


# =============================================================================
# Comparison logic
# =============================================================================

def main(xlsx_path):
    wb = openpyxl.load_workbook(xlsx_path, data_only=True)

    # ---- Meeting-level sheet ----
    ws_m = wb['meeting_level']
    meeting_xlsx = {}
    for r in range(2, ws_m.max_row + 1):
        start = ws_m.cell(r, 2).value  # dd/mm/yyyy
        if start is None:
            continue
        meeting_xlsx[start] = {
            "res_rate":   ws_m.cell(r, 4).value,
            "res_action": ws_m.cell(r, 5).value,
            "res_stance": ws_m.cell(r, 6).value,
        }

    # ---- Member-level sheet ----
    ws = wb['member_level_long']
    xlsx_data = {}
    for r in range(2, ws.max_row + 1):
        start = ws.cell(r, 2).value
        name = ws.cell(r, 5).value
        if start is None or name is None:
            continue
        xlsx_data.setdefault(start, {"members": {}})
        xlsx_data[start]["members"][name] = {
            "rate_vote":        ws.cell(r, 8).value,
            "stance_vote":      ws.cell(r, 9).value,
            "preferred_stance": ws.cell(r, 14).value,
            "delta_bps":        ws.cell(r, 21).value,
            "impl_rate":        ws.cell(r, 22).value,
        }

    discrepancies = []

    # Compare GT vs xlsx
    for date_key, gt in sorted(GT.items()):
        # Meeting-level checks
        if date_key not in meeting_xlsx:
            discrepancies.append(f"{date_key} META: meeting missing in 'meeting_level' sheet")
        else:
            xm = meeting_xlsx[date_key]
            if xm["res_rate"] is None or abs(float(xm["res_rate"]) - gt["rate"]) > 0.001:
                discrepancies.append(
                    f"{date_key} META: rate mismatch xlsx={xm['res_rate']} gt={gt['rate']}"
                )
            if xm["res_action"] != gt["rate_action"]:
                discrepancies.append(
                    f"{date_key} META: rate_action mismatch xlsx={xm['res_action']} gt={gt['rate_action']}"
                )
            if xm["res_stance"] != gt["stance"]:
                discrepancies.append(
                    f"{date_key} META: stance mismatch xlsx={xm['res_stance']} gt={gt['stance']}"
                )

        # Member-level checks
        if date_key not in xlsx_data:
            discrepancies.append(
                f"{date_key} MEMBER-SHEET: meeting missing in 'member_level_long'"
            )
            continue

        xd = xlsx_data[date_key]
        for member, mgt in gt["members"].items():
            if member not in xd["members"]:
                discrepancies.append(f"{date_key} MEMBER MISSING: {member}")
                continue
            mxd = xd["members"][member]

            # rate_vote
            if str(mxd["rate_vote"]).strip() != mgt["rate_vote"]:
                discrepancies.append(
                    f"{date_key} {member}: rate_vote xlsx={mxd['rate_vote']!r} gt={mgt['rate_vote']!r}"
                )

            # stance_vote
            if str(mxd["stance_vote"]).strip() != mgt["stance_vote"]:
                discrepancies.append(
                    f"{date_key} {member}: stance_vote xlsx={mxd['stance_vote']!r} gt={mgt['stance_vote']!r}"
                )

            # preferred_stance (normalise None / 'None')
            xpref = mxd["preferred_stance"]
            if xpref is not None:
                xpref = str(xpref).strip()
                if xpref.lower() in {"none", ""}:
                    xpref = None
            if xpref != mgt["preferred_stance"]:
                discrepancies.append(
                    f"{date_key} {member}: preferred_stance "
                    f"xlsx={xpref!r} gt={mgt['preferred_stance']!r}"
                )

            # delta_bps
            xdelta = mxd["delta_bps"]
            gdelta = mgt["delta_bps"]
            if xdelta is None:
                if gdelta != 0:
                    discrepancies.append(
                        f"{date_key} {member}: delta_bps xlsx=None gt={gdelta}"
                    )
            else:
                try:
                    if int(xdelta) != gdelta:
                        discrepancies.append(
                            f"{date_key} {member}: delta_bps xlsx={xdelta} gt={gdelta}"
                        )
                except (ValueError, TypeError):
                    discrepancies.append(
                        f"{date_key} {member}: delta_bps unparseable xlsx={xdelta!r} gt={gdelta}"
                    )

        # Extra members in xlsx not in GT
        for m in xd["members"]:
            if m not in gt["members"]:
                discrepancies.append(f"{date_key} EXTRA MEMBER in xlsx (not in GT): {m}")

    # ---- Report ----
    total_meetings = len(GT)
    total_member_rows = sum(len(g["members"]) for g in GT.values())

    print("=" * 70)
    print(f"Cross-check of: {xlsx_path}")
    print("=" * 70)
    print(f"Ground-truth coverage: {total_meetings} meetings, "
          f"{total_member_rows} member-rows")
    print()

    if not discrepancies:
        print("RESULT: PASS — no discrepancies found.")
        return 0

    print(f"RESULT: {len(discrepancies)} discrepancy/discrepancies found.")
    print("-" * 70)
    for d in discrepancies:
        print(d)
    return 1


main(XLSX_PATH)
```
